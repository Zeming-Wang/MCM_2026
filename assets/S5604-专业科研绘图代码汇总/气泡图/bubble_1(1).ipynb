### **气泡图主要构成**

气泡图是一种用于显示数据关系和分布的图形，能够同时展示三维（X、Y轴和气泡大小）甚至四维数据（再加上气泡颜色）之间的关联。它的关键构成元素包括：

- **横轴和纵轴**：这两者分别表示两个不同变量的数值，可以是数值型、分类型或时间型数据。
- **气泡大小**：用于表示第三个变量的数值。气泡的尺寸通常与该变量的值成正比，数值越大，气泡越大。
- **气泡颜色**：颜色可以作为第四个维度来使用，用于展示另一变量的变化。不同颜色通常用于区分数据类别或表示数值范围的不同。

这种图形可以同时展示多个维度的数据，适合于展示复杂的关系。

`plt.scatter()` 是 `matplotlib` 中用于绘制散点图的函数。通过 X 和 Y 轴的数据绘制点，同时还能自定义点的大小、颜色、形状等。

### 函数签名
```python
matplotlib.pyplot.scatter(x, y, s=None, c=None, marker=None, cmap=None, norm=None, vmin=None, vmax=None, alpha=None, linewidths=None, edgecolors=None, plotnonfinite=False, data=None, **kwargs)
```

### 参数详解

1. **`x`**: 
   - 说明：表示 X 轴数据。
   - 类型：数组或列表。
   - 必须参数。

2. **`y`**:
   - 说明：表示 Y 轴数据。
   - 类型：数组或列表。
   - 必须参数。

3. **`s`**:
   - 说明：每个散点的大小。
   - 类型：标量或数组，默认值为 20。通常大小与数据相关时，传入一个与 `x` 和 `y` 等长度的数组。
   - 使用示例：`s=[30, 60, 90]`。

4. **`c`**:
   - 说明：散点的颜色，可以是单一颜色或每个点不同颜色。可以传入颜色值数组，也可以传入一个表示数据值的数组，并结合 `cmap` 来生成颜色。
   - 类型：标量、字符串、数组或 `None`。
   - 使用示例：`c='blue'` 或 `c=[1, 2, 3]`。

5. **`marker`**:
   - 说明：表示散点的形状。
   - 类型：字符串，表示形状的符号。
   - 使用示例：`marker='o'`（圆形），`marker='s'`（方形），`marker='^'`（三角形），更多符号见 [Matplotlib Markers](https://matplotlib.org/stable/api/markers_api.html)。

6. **`cmap`**:
   - 说明：颜色映射（当 `c` 参数是数值时生效），用于将数值映射为颜色。
   - 类型：`Colormap` 实例或字符串，默认 `None`。
   - 使用示例：`cmap='viridis'`。

7. **`norm`**:
   - 说明：用于将 `c` 的值标准化到 `[0, 1]` 的范围，通常与 `cmap` 配合使用。
   - 类型：`Normalize` 实例或 `None`。
   - 使用示例：`norm=plt.Normalize(vmin=0, vmax=100)`。

8. **`vmin` 和 `vmax`**:
   - 说明：用于设置 `c` 值的最小值和最大值。如果 `c` 是数值，`vmin` 和 `vmax` 控制色图的范围。
   - 类型：标量或 `None`，默认 `None`。
   - 使用示例：`vmin=0, vmax=100`。

9. **`alpha`**:
   - 说明：控制散点的透明度，取值范围 `[0, 1]`，`1` 表示完全不透明。
   - 类型：标量或 `None`，默认 `None`（完全不透明）。
   - 使用示例：`alpha=0.7`。

10. **`linewidths`**:
    - 说明：每个散点的边框线宽。如果没有设置边框颜色，则忽略此参数。
    - 类型：标量或数组，默认 `None`。
    - 使用示例：`linewidths=2`。

11. **`edgecolors`**:
    - 说明：设置散点的边框颜色。可以使用单个颜色或与数据长度相同的数组。默认值为 `'face'`，即边框颜色与点的填充颜色相同。
    - 类型：字符串或颜色序列，默认 `'face'`。
    - 使用示例：`edgecolors='black'` 或 `edgecolors=['red', 'green', 'blue']`。

12. **`plotnonfinite`**:
    - 说明：如果为 `True`，则绘制 `NaN` 或无穷大的值为散点。默认为 `False`。
    - 类型：布尔值，默认 `False`。

13. **`data`**:
    - 说明：数据参数，用于从 `DataFrame` 等结构中获取绘图数据。
    - 类型：`dict` 或 `DataFrame`，默认 `None`。

### 其他参数（`**kwargs`）
可以通过 `**kwargs` 传递其他 `matplotlib` 图形对象的参数，例如 `label`、`zorder` 等。

### **下面是几个基础示例**

````python
import matplotlib.pyplot as plt
   
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.scatter(x, y)
plt.show()
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
sizes = [100, 200, 300, 400]  # 气泡大小
colors = ['red', 'blue', 'green', 'purple']  # 颜色

plt.scatter(x, y, s=sizes, c=colors, alpha=0.5)
plt.show()
````

````python
import numpy as np
   
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, c=colors, cmap='viridis', s=100, vmin=0, vmax=1)
plt.colorbar()  # 添加颜色条
plt.show()
````

## **区域、人口、亚洲人口比例分布图**

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np

# "percasian"这个术语通常出现在涉及人口统计或社会学研究的上下文中，在分析人口构成时。它代表了一个地区或群体中亚洲人（Asian）的比例或百分比。
# 读取数据
data = pd.read_csv('midwest_filter.csv')

# 自定义图形参数
plt.rcParams.update({
    'figure.figsize': (10, 6),   # 调整画布尺寸
    'axes.titlesize': 20,        # 设置标题字体大小
    'axes.labelsize': 14,        # 轴标签字体大小
    'xtick.labelsize': 12,       # X轴刻度字体大小
    'ytick.labelsize': 12,       # Y轴刻度字体大小
    'legend.fontsize': 12,       # 图例字体大小
})

# 使用 seaborn 风格
sns.set_theme(style="whitegrid")

# 类别和颜色生成
groups = data['category'].unique()
color_palette = sns.color_palette("Set2", len(groups))

# 初始化画布
plt.figure(facecolor='lightblue')

# 绘制散点图，不同类别使用不同颜色
for idx, grp in enumerate(groups):
    subset = data[data['category'] == grp]
    plt.scatter(
        subset['area'], 
        subset['poptotal'], 
        s=subset['percasian'] * 600,  # 点大小与 percasian 成比例
        color=color_palette[idx], 
        label=grp, 
        edgecolor='k', 
        alpha=0.6  # 设置透明度
    )

# 设置坐标轴标签、标题及范围
plt.xlabel('Area of Region')
plt.ylabel('Total Population')
plt.title('Population vs Area by Category', fontsize=20)

# 添加图例
plt.legend(title='Category', loc='upper right')

# 限制坐标轴范围
plt.xlim(0, 0.12)
plt.ylim(0, 90000)
plt.grid(False)
# 显示图像
plt.tight_layout()  # 调整布局避免重叠
plt.savefig('population_bubble.png', dpi=300)
plt.show()
````

## **不同年龄、性别、收入、购买力比较图**

````python
import matplotlib.pyplot as plt
import pandas as pd

# 从CSV文件中读取数据
df = pd.read_csv('Mall_Customers_Int.csv')

# 绘制气泡图
plt.figure(figsize=(10, 6))

# 根据性别绘制不同的气泡
# 性别为1（假设为男性）
male = df[df['Genre'] == 1]
plt.scatter(male['Age'], male['Annual_Income_(k$)'], 
            s=male['Spending_Score']*10,  # 根据消费分数调整气泡大小
            c='#ff9400', label='Male', alpha=0.6, linewidth=1)

# 性别为0（假设为女性）
female = df[df['Genre'] == 0]
plt.scatter(female['Age'], female['Annual_Income_(k$)'], 
            s=female['Spending_Score']*10,  # 根据消费分数调整气泡大小
            c='#7600ff', label='Female', alpha=0.6, linewidth=1)

# 取消网格
plt.grid(False)

# 设置坐标轴标签和标题
plt.xlabel('Age')
plt.ylabel('Annual Income (k$)')
plt.title('Bubble Chart of Customer Data')

# 显示图例
plt.legend(title='Gender')
plt.savefig('Mall_Customers_bubble.png', dpi=300)
# 显示图像
plt.show()
````
