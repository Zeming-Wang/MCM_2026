```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
x = np.random.normal(50, 15, 100)  # x 轴数据
y = np.random.normal(50, 15, 100)  # y 轴数据
sizes = np.random.uniform(100, 1000, 100)  # 气泡大小
colors = np.random.rand(100)  # 气泡颜色（随机值）

# 创建画布
fig = plt.figure(figsize=(16, 10))

# 主气泡图
ax_main = fig.add_axes([0.1, 0.3, 0.6, 0.6])  # 手动指定位置和大小
scatter = ax_main.scatter(
    x, y,
    s=sizes,  # 气泡大小
    c=colors,  # 气泡颜色
    cmap='viridis',  # 颜色映射
    alpha=0.8,  # 透明度
    edgecolor='black',  # 气泡边缘颜色
    linewidth=0.5  # 边缘线宽
)
ax_main.set_xlabel('X 轴', fontsize=12)
ax_main.set_ylabel('Y 轴', fontsize=12)
ax_main.set_title('气泡图与分布情况', fontsize=16, pad=20)

# 添加颜色条
cbar = plt.colorbar(scatter, ax=ax_main, orientation='vertical', pad=0.07)
cbar.set_label('颜色值', fontsize=12)

# 只显示 x 轴和 y 轴的线，隐藏其他边框线
ax_main.spines['top'].set_visible(False)    # 隐藏上边框
ax_main.spines['right'].set_visible(False)  # 隐藏右边框
ax_main.spines['bottom'].set_visible(True)  # 显示下边框（x 轴）
ax_main.spines['left'].set_visible(True)    # 显示左边框（y 轴）

# X 轴分布条形码（调整到下边）
ax_x_dist = fig.add_axes([0.10, 0.2, 0.47, 0.05])  # 手动指定位置和大小
ax_x_dist.eventplot(x, orientation='horizontal', colors='#0088ba', lineoffsets=0.5, linelengths=0.8)
ax_x_dist.set_xlim(ax_main.get_xlim())  # 与主图 x 轴范围一致
ax_x_dist.axis('off')  # 隐藏坐标轴

# Y 轴分布条形码（调整到左边，自定义位置）
ax_y_dist = fig.add_axes([0.03, 0.3, 0.035, 0.6])  # 手动指定位置和大小
ax_y_dist.eventplot(y, orientation='vertical', colors='#0088ba', lineoffsets=0.5, linelengths=0.8)
ax_y_dist.set_ylim(ax_main.get_ylim())  # 与主图 y 轴范围一致
ax_y_dist.axis('off')  # 隐藏坐标轴

# 显示图形
plt.savefig('bubble_plot.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C']  # 分组名称
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # 每组颜色
markers = ['o', 'o', 'o']  # 每组标记

# 生成每组的数据
x = []
y = []
sizes = []  # 气泡大小数据
for i, group in enumerate(groups):
    x.append(np.random.normal(50 + i * 10, 15, 100))  # x 轴数据
    y.append(np.random.normal(50 + i * 10, 15, 100))  # y 轴数据
    sizes.append(np.random.uniform(10, 300, 100))  # 气泡大小数据

# 创建画布
fig = plt.figure(figsize=(14, 10))

# 主气泡图
ax_main = fig.add_axes([0.1, 0.3, 0.6, 0.6])  # 手动指定位置和大小

# 绘制分组气泡图
for i, group in enumerate(groups):
    ax_main.scatter(
        x[i], y[i],
        color=colors[i],  # 气泡颜色
        marker=markers[i],  # 气泡形状
        alpha=0.8,  # 透明度
        edgecolor=colors[i],  # 边缘颜色
        s=sizes[i],  # 气泡大小
        label=group  # 分组标签
    )

# 设置坐标轴的颜色和线型
for spine in ax_main.spines.values():
    spine.set_color('#1f77b4')  # 设置坐标轴线颜色为蓝色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 设置刻度线的颜色和线型
ax_main.tick_params(axis='both', which='both', color='#1f77b4', width=2, length=6, direction='inout')

# 添加图例
ax_main.legend(loc='upper right', fontsize=12)

# 设置标题和轴标签
ax_main.set_xlabel('X 轴', fontsize=12, color='#1f77b4')
ax_main.set_ylabel('Y 轴', fontsize=12, color='#1f77b4')
ax_main.set_title('分组气泡图', fontsize=16, pad=20, color='#1f77b4')

# X 轴分布条形码（调整到下边，合并绘制）
ax_x_dist = fig.add_axes([0.1, 0.18, 0.6, 0.08])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_x_dist.eventplot(x[i], orientation='horizontal', colors=colors[i], lineoffsets=0.5, linelengths=0.2)
ax_x_dist.set_xlim(ax_main.get_xlim())  # 与主图 x 轴范围一致
ax_x_dist.axis('off')  # 隐藏坐标轴

# Y 轴分布条形码（调整到左边，合并绘制）
ax_y_dist = fig.add_axes([0.01, 0.3, 0.05, 0.6])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_y_dist.eventplot(y[i], orientation='vertical', colors=colors[i], lineoffsets=0.5, linelengths=0.2)
ax_y_dist.set_ylim(ax_main.get_ylim())  # 与主图 y 轴范围一致
ax_y_dist.axis('off')  # 隐藏坐标轴

# 显示图形
plt.savefig('bubble_group.png', dpi=400, bbox_inches='tight')
plt.show()
```

# 六轴气泡图

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C', 'Group D', 'Group E', 'Group F']  # 六组名称
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']  # 每组颜色

# 生成每组的数据
data = []
sizes = []  # 气泡大小数据

for i in range(len(groups)):
    data.append(np.random.normal(50 + 5 * np.random.randint(1, 5), 10, 25))  # 每组生成 20 个正态分布的数据点
    sizes.append(np.random.uniform(10, 600, 25))  # 气泡大小数据

# 创建画布
fig, ax = plt.subplots(figsize=(10, 8))

# 绘制六组单轴散点图
for i, group in enumerate(groups):
    ax.scatter(
        data[i],  # x 轴数据
        np.full_like(data[i], i),  # y 轴数据（每组数据沿不同的 y 轴位置分布）
        color=colors[i],  # 数据点颜色
        alpha=0.8,  # 透明度
        edgecolor=colors[i],  # 边缘颜色
        s=sizes[i],  # 数据点大小
        label=group  # 分组标签
    )

# 在每个轴位置绘制一条虚线
for i in range(len(groups)):
    ax.axhline(
        y=i,  # y 轴位置
        color='gray',  # 虚线颜色
        linestyle='--',  # 虚线样式
        linewidth=1,  # 线宽
        alpha=0.5  # 透明度
    )

# 设置 y 轴刻度和标签
ax.set_yticks(range(len(groups)))  # 设置 y 轴刻度
ax.set_yticklabels(groups, fontsize=12)  # 设置 y 轴标签

# 设置标题和轴标签
ax.set_xlabel('数据值', fontsize=12, color='#1f77b4')
ax.set_ylabel('分组', fontsize=12, color='#1f77b4')
ax.set_title('六组单轴气泡图（带虚线）', fontsize=16, pad=20, color='#1f77b4')

# 设置坐标轴的颜色和线型
# for spine in ax.spines.values():
#     spine.set_color('#1f77b4')  # 设置坐标轴线颜色为蓝色
#     spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
#     spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 只显示 x 轴和 y 轴的线，隐藏其他边框线
ax.spines['top'].set_visible(False)    # 隐藏上边框
ax.spines['right'].set_visible(False)  # 隐藏右边框
ax.spines['bottom'].set_visible(True)  # 显示下边框（x 轴）
ax.spines['bottom'].set_color('#ffaba5')  # 显示下边框（x 轴）
ax.spines['left'].set_visible(False)    # 隐藏左边框（y 轴）
ax.tick_params(colors='#ffaba5')
# 添加图例
ax.legend(loc='upper right', fontsize=12)

# 显示图形
plt.savefig('six_group_single_axis_bubble_with_lines.png', dpi=400, bbox_inches='tight')
plt.show()
```
