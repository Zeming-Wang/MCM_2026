# 基本饼图

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体  
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号
# 示例数据
sizes = [15, 30, 45, 10]  # 各扇区的占比
labels = ['A', 'B', 'C', 'D']  # 各扇区的标签
explode = (0, 0.1, 0, 0)  # 突出显示第二个扇区
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']  # 自定义颜色

# 创建画布和子图
fig, ax = plt.subplots()

# 绘制饼图
wedges, texts, autotexts = ax.pie(
    sizes,  # 扇区大小
    explode=explode,  # 突出显示某个扇区
    labels=labels,  # 扇区标签
    colors=colors,  # 扇区颜色
    autopct='%1.1f%%',  # 显示百分比
    startangle=90,  # 起始角度
    shadow=True,  # 添加阴影
    labeldistance=1.2,  # 标签距离圆心的距离
    pctdistance=0.8,  # 百分比标签距离圆心的距离
    wedgeprops={'edgecolor': 'black', 'linewidth': 2},  # 扇区边框样式
    textprops={'fontsize': 12, 'color': 'black'},  # 文本样式
    center=(0, 0),  # 饼图中心位置
    radius=1,  # 饼图半径
    counterclock=True,  # 扇区绘制方向（逆时针）
    rotatelabels=False,  # 是否旋转标签
    normalize=True,  # 是否归一化数据
)

# 设置标题
ax.set_title("饼图示例", fontsize=16)
plt.savefig("基本饼图.png", dpi=300)
# 显示图形
plt.show()
```

### 添加条形图示例

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import ConnectionPatch

# 创建画布和子图
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
fig.subplots_adjust(wspace=0)  # 调整子图间距

# 饼图参数
overall_ratios = [0.27, 0.56, 0.17]  # 各扇区的占比
labels = ['Approve', 'Disapprove', 'Undecided']  # 扇区标签
explode = [0.1, 0, 0]  # 突出显示第一个扇区
angle = -180 * overall_ratios[0]  # 旋转角度，使第一个扇区从 x 轴开始

# 绘制饼图
wedges, texts, autotexts = ax1.pie(
    overall_ratios,
    explode=explode,
    labels=labels,
    autopct='%1.1f%%',
    startangle=angle,
    colors=['#ff9999', '#66b3ff', '#99ff99'],  # 自定义颜色
    wedgeprops={'edgecolor': 'black', 'linewidth': 2},  # 扇区边框样式
    textprops={'fontsize': 12, 'color': 'black'},  # 文本样式
)

# 条形图参数
age_ratios = [0.33, 0.54, 0.07, 0.06]  # 各年龄段的占比
age_labels = ['Under 35', '35-49', '50-65', 'Over 65']  # 年龄段标签
bottom = 1  # 条形图的起始位置
width = 0.2  # 条形图的宽度

# 绘制条形图
for j, (height, label) in enumerate(reversed([*zip(age_ratios, age_labels)])):
    bottom -= height  # 计算每个条形图的底部位置
    bc = ax2.bar(
        0, height, width, bottom=bottom, color='C0', label=label,
        alpha=0.1 + 0.25 * j  # 设置透明度
    )
    ax2.bar_label(bc, labels=[f"{height:.0%}"], label_type='center')  # 添加百分比标签

# 设置条形图标题和图例
ax2.set_title('Age of Approvers', fontsize=14)
ax2.legend(loc='upper right')
ax2.axis('off')  # 隐藏坐标轴
ax2.set_xlim(-2.5 * width, 2.5 * width)  # 设置 x 轴范围

# 使用 ConnectionPatch 绘制连接线
theta1, theta2 = wedges[0].theta1, wedges[0].theta2  # 获取第一个扇区的起始和结束角度
center, r = wedges[0].center, wedges[0].r  # 获取饼图的中心和半径
bar_height = sum(age_ratios)  # 条形图的总高度

# 绘制顶部连接线
x = r * np.cos(np.pi / 180 * theta2) + center[0]  # 计算连接点的 x 坐标
y = r * np.sin(np.pi / 180 * theta2) + center[1]  # 计算连接点的 y 坐标
con = ConnectionPatch(
    xyA=(-width / 2, bar_height), coordsA=ax2.transData,  # 条形图的连接点
    xyB=(x, y), coordsB=ax1.transData,  # 饼图的连接点
    color='black', linewidth=2  # 连接线样式
)
ax2.add_artist(con)

# 绘制底部连接线
x = r * np.cos(np.pi / 180 * theta1) + center[0]  # 计算连接点的 x 坐标
y = r * np.sin(np.pi / 180 * theta1) + center[1]  # 计算连接点的 y 坐标
con = ConnectionPatch(
    xyA=(-width / 2, 0), coordsA=ax2.transData,  # 条形图的连接点
    xyB=(x, y), coordsB=ax1.transData,  # 饼图的连接点
    color='black', linewidth=2  # 连接线样式
)
ax2.add_artist(con)
plt.savefig("添加条形图饼图.png", dpi=300)
# 显示图形
plt.show()
```

## 设置图例和标签

```python
import matplotlib.pyplot as plt
import numpy as np

# 创建画布和子图
def create_figure():
    """创建画布和子图"""
    fig, ax = plt.subplots(figsize=(8, 4), subplot_kw=dict(aspect="equal"))
    return fig, ax

# 示例数据：每日时间分配
def get_daily_activities():
    """返回每日时间分配的数据"""
    activities = ["Sleep", "Work", "Leisure", "Exercise", "Meals"]
    time_spent = [8, 9, 4, 1, 2]  # 单位：小时
    return activities, time_spent

# 自定义百分比标签格式
def format_percentage(pct, allvals):
    """
    自定义百分比标签格式，显示百分比和实际值。
    :param pct: 百分比值
    :param allvals: 所有数据的值
    :return: 格式化后的字符串
    """
    absolute = int(np.round(pct / 100 * np.sum(allvals)))  # 计算实际值
    return f"{pct:.1f}%\n({absolute:d} h)"  # 返回格式化字符串

# 绘制饼图
def plot_pie_chart(ax, data, labels, colors):
    """
    绘制饼图
    :param ax: 子图对象
    :param data: 数据值
    :param labels: 数据标签
    :param colors: 颜色列表
    """
    wedges, texts, autotexts = ax.pie(
        data,
        autopct=lambda pct: format_percentage(pct, data),  # 使用自定义标签格式
        textprops=dict(color="w"),  # 设置文本颜色为白色
        colors=colors,  # 自定义颜色
        wedgeprops={'edgecolor': 'black', 'linewidth': 1},  # 扇区边框样式
        startangle=90,  # 起始角度
    )
    return wedges, autotexts

# 添加图例
def add_legend(ax, wedges, labels):
    """
    添加图例
    :param ax: 子图对象
    :param wedges: 饼图扇区对象
    :param labels: 图例标签
    """
    ax.legend(
        wedges, labels,  # 图例内容和标签
        title="Activities",  # 图例标题
        loc="center left",  # 图例位置
        bbox_to_anchor=(1, 0, 0.5, 1),  # 图例位置微调
    )

# 设置百分比标签样式
def set_autotext_style(autotexts):
    """
    设置百分比标签样式
    :param autotexts: 百分比标签对象
    """
    plt.setp(autotexts, size=10, weight="bold")  # 设置字体大小和加粗

# 设置标题
def set_title(ax, title):
    """
    设置标题
    :param ax: 子图对象
    :param title: 标题内容
    """
    ax.set_title(title, fontsize=14)

# 主函数
def main():
    # 创建画布和子图
    fig, ax = create_figure()

    # 获取数据
    activities, time_spent = get_daily_activities()

    # 定义颜色
    colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

    # 绘制饼图
    wedges, autotexts = plot_pie_chart(ax, time_spent, activities, colors)

    # 添加图例
    add_legend(ax, wedges, activities)

    # 设置百分比标签样式
    set_autotext_style(autotexts)

    # 设置标题
    set_title(ax, "Daily Time Allocation")

    # 显示图形
    plt.tight_layout()  # 调整布局
    plt.savefig("设置图例和标签.png", dpi=300)
    plt.show()

# 运行主函数
if __name__ == "__main__":
    main()
```

# 嵌套环形图

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体  
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号
# 创建画布和子图
fig, ax = plt.subplots(figsize=(8, 8))

# 设置嵌套饼图的参数
size = 0.3  # 内层饼图的宽度

# 示例数据：公司年度销售数据
# 外层：三个地区的销售总额
# 内层：每个地区的三种产品类别的销售数据
vals = np.array([
    [60, 32, 25],  # 地区 A：产品 1、产品 2、产品 3
    [37, 40, 20],  # 地区 B：产品 1、产品 2、产品 3
    [29, 10, 15],  # 地区 C：产品 1、产品 2、产品 3
])

# Pastel 柔和色系
outer_colors = ['#A6CEE3', '#B2DF8A', '#FB9A99']  # 外层颜色：浅蓝、浅绿、浅橙
inner_colors = [
    '#C6DBEF', '#D9F0A3', '#FCAE91',  # 地区 A 的产品颜色
    '#9ECAE1', '#A1D99B', '#FC9272',  # 地区 B 的产品颜色
    '#6BAED6', '#74C476', '#FB6A4A',  # 地区 C 的产品颜色
]

# 绘制外层饼图（地区销售总额）
ax.pie(
    vals.sum(axis=1),  # 计算每个地区的销售总额
    radius=1,  # 外层饼图的半径
    colors=outer_colors,  # 外层颜色
    wedgeprops=dict(width=size, edgecolor='w', linewidth=2),  # 扇区样式
    labels=['地区 A', '地区 B', '地区 C'],  # 地区标签
    startangle=90,  # 起始角度
    textprops={'fontsize': 12, 'color': 'black'},  # 文本样式
)

# 绘制内层饼图（产品类别销售数据）
ax.pie(
    vals.flatten(),  # 将数据展平为一维数组
    radius=1 - size,  # 内层饼图的半径
    colors=inner_colors,  # 内层颜色
    wedgeprops=dict(width=size, edgecolor='w', linewidth=2),  # 扇区样式
    startangle=90,  # 起始角度
    labels=[
        '地区 A - 产品 1', '地区 A - 产品 2', '地区 A - 产品 3',
        '地区 B - 产品 1', '地区 B - 产品 2', '地区 B - 产品 3',
        '地区 C - 产品 1', '地区 C - 产品 2', '地区 C - 产品 3',
    ],  # 产品类别标签
    labeldistance=0.75,  # 标签距离圆心的距离
    textprops={'fontsize': 10, 'color': 'black'},  # 文本样式
)

# 设置标题
ax.set_title("各地区及产品类别年度销售分布", fontsize=14, pad=20)

# 显示图形
plt.tight_layout()
plt.savefig("环形图.png", dpi=300)
plt.show()
```

# 三层嵌套环形图

```python
import matplotlib.pyplot as plt

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 圆环宽度
RING_WIDTH = 0.4

# 示例数据
data = [[[20, 30], 32], [37, [20, 15]], [29, 10, [10, 5]]]

# 数据处理函数
def process_data(data):
    """
    处理嵌套数据，返回内层、中层和外层的数据列表。
    :param data: 嵌套数据
    :return: 内层数据、中层数据、外层数据
    """
    inner_data = []  # 内层数据（一级分组合计）
    middle_data = []  # 中层数据（二级分组合计）
    outer_data = []  # 外层数据（原始数据）

    for group in data:
        group_sum = 0  # 一级分组合计
        temp_middle_data = []  # 临时存储二级分组数据

        for item in group:
            if isinstance(item, list):  # 如果二级分组是列表
                group_sum += sum(item)
                temp_middle_data.append(sum(item))
                outer_data.extend(item)
            else:  # 如果二级分组是单个值
                group_sum += item
                temp_middle_data.append(item)
                outer_data.append(item)

        inner_data.append(group_sum)
        middle_data.extend(temp_middle_data)

    return inner_data, middle_data, outer_data

# 绘制圆环图
def plot_ring(data, radius, pct_distance, label=None):
    """
    绘制圆环图。
    :param data: 数据列表
    :param radius: 圆环半径
    :param pct_distance: 百分比标签距离圆心的距离
    :param label: 图例标签
    """
    plt.pie(
        data,
        radius=radius,
        autopct='%1.1f%%',
        pctdistance=pct_distance,
        wedgeprops=dict(width=RING_WIDTH, edgecolor='w'),
        labels=[label] * len(data) if label else None,
    )

# 处理数据
inner_data, middle_data, outer_data = process_data(data)

# 打印数据
print("外层数据（原始数据）:", outer_data)
print("中层数据（二级分组合计）:", middle_data)
print("内层数据（一级分组合计）:", inner_data)

# 绘制外层圆环（原始数据）
plot_ring(outer_data, radius=1.4, pct_distance=0.9)

# 绘制中层圆环（二级分组合计）
plot_ring(middle_data, radius=1.4 - RING_WIDTH, pct_distance=0.8)

# 绘制内层圆环（一级分组合计）
plot_ring(inner_data, radius=1.4 - 2 * RING_WIDTH, pct_distance=0.6)

# 设置标题
plt.title('三层圆环图', loc='left', fontsize=14, pad=20)
plt.savefig("三层环形图.png", dpi=400)
# 显示图形
plt.show()
```

# 设置颜色

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 圆环宽度
RING_WIDTH = 0.4

# 示例数据
data = [
    [[30, 40], 50],  # 第一组
    [60, [25, 35]],  # 第二组
    [45, 20, [15, 10]]  # 第三组
]

# 数据处理函数
def process_data(data):
    """
    处理嵌套数据，返回内层、中层和外层的数据列表。
    :param data: 嵌套数据
    :return: 内层数据、中层数据、外层数据
    """
    inner_data = []  # 内层数据（一级分组合计）
    middle_data = []  # 中层数据（二级分组合计）
    outer_data = []  # 外层数据（原始数据）

    for group in data:
        group_sum = 0  # 一级分组合计
        temp_middle_data = []  # 临时存储二级分组数据

        for item in group:
            if isinstance(item, list):  # 如果二级分组是列表
                group_sum += sum(item)
                temp_middle_data.append(sum(item))
                outer_data.extend(item)
            else:  # 如果二级分组是单个值
                group_sum += item
                temp_middle_data.append(item)
                outer_data.append(item)

        inner_data.append(group_sum)
        middle_data.extend(temp_middle_data)

    return inner_data, middle_data, outer_data

# 绘制圆环图
def plot_ring(data, radius, pct_distance, colors, label=None):
    """
    绘制圆环图。
    :param data: 数据列表
    :param radius: 圆环半径
    :param pct_distance: 百分比标签距离圆心的距离
    :param colors: 颜色列表
    :param label: 图例标签
    """
    plt.pie(
        data,
        radius=radius,
        autopct='%1.1f%%',
        pctdistance=pct_distance,
        wedgeprops=dict(width=RING_WIDTH, edgecolor='w'),
        labels=[label] * len(data) if label else None,
        colors=colors,
        textprops={'color': 'black', 'fontsize': 10}  # 设置百分比标签的颜色和字体大小
    )

# 处理数据
inner_data, middle_data, outer_data = process_data(data)

# 打印数据
print("外层数据（原始数据）:", outer_data)
print("中层数据（二级分组合计）:", middle_data)
print("内层数据（一级分组合计）:", inner_data)

# 生成渐变色
def generate_colors(n, colormap='viridis', start=0.2, end=0.8):
    """
    生成渐变色列表，避免生成过深的颜色。
    :param n: 颜色数量
    :param colormap: 颜色映射名称
    :param start: 颜色映射起始位置（0到1之间）
    :param end: 颜色映射结束位置（0到1之间）
    :return: 颜色列表
    """
    cmap = plt.get_cmap(colormap)
    return [cmap(i) for i in np.linspace(start, end, n)]

# 生成颜色
outer_colors = generate_colors(len(outer_data), 'plasma', start=0.2, end=0.8)
middle_colors = generate_colors(len(middle_data), 'viridis', start=0.2, end=0.8)
inner_colors = generate_colors(len(inner_data), 'cool', start=0.2, end=0.8)

# 绘制外层圆环（原始数据）
plot_ring(outer_data, radius=1.4, pct_distance=0.9, colors=outer_colors)

# 绘制中层圆环（二级分组合计）
plot_ring(middle_data, radius=1.4 - RING_WIDTH, pct_distance=0.8, colors=middle_colors)

# 绘制内层圆环（一级分组合计）
plot_ring(inner_data, radius=1.4 - 2 * RING_WIDTH, pct_distance=0.6, colors=inner_colors)

# 设置标题
plt.title('三层圆环图', loc='left', fontsize=14, pad=20)

# 显示图形
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

# 常量配置
FONT_PATH = r'../simkai.ttf'  # 字体路径
DATA_FILE = '疫情各国确诊数据.csv'
OUTPUT_IMAGE = '南丁格尔图.png'
CHART_TITLE = '感染累计'
FIGURE_SIZE = (10, 6)
DPI = 400

# 设置字体
def setup_font(font_path):
    """设置中文字体"""
    font_prop = fm.FontProperties(fname=font_path)
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False

# 加载数据
def load_data(file_path):
    """加载数据并处理"""
    df = pd.read_csv(file_path)
    head_df = df.head(24).iloc[1:]  # 取前24个，去掉美国
    head_df.sort_values(by='确诊病例', ascending=False, inplace=True)  # 降序排序
    return head_df['国家/地区'].values.tolist(), head_df['确诊病例'].values.tolist()

# 绘制极坐标图
def draw_polar_chart(labels, data):
    """绘制极坐标图"""
    fig = plt.figure(figsize=FIGURE_SIZE, dpi=DPI)
    ax = plt.axes(polar=True)
    ax.set_theta_zero_location("N")  # 极坐标 0° 方向为 N
    ax.set_theta_direction(-1)  # 顺时针为极坐标正方向
    theta = np.linspace(0, np.pi * 2, len(data), endpoint=False)  # 等分极坐标系

    # 绘制柱状图
    ax.bar(
        x=theta,
        height=data,
        width=np.pi / len(data) * 2,
        color=np.random.random((len(data), 3)),
        align="edge"
    )

    # 绘制中心空白
    ax.bar(
        x=theta,
        height=25000,
        width=np.pi / len(data) * 2,
        color='white'
    )

    # 设置标题
    ax.set_title(CHART_TITLE, fontdict={'fontsize': 20})

    # 标注文本
    for angle, height, text in zip(theta, data, labels):
        ax.text(angle + 0.25, height + 120, str(text), fontsize=5)

    plt.axis("off")
    plt.savefig(OUTPUT_IMAGE, dpi=400)
    plt.show()

def main():
    """主函数"""
    setup_font(FONT_PATH)
    labels, data = load_data(DATA_FILE)
    draw_polar_chart(labels, data)

if __name__ == '__main__':
    main()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import funcy
import warnings
import matplotlib.font_manager as fm

# 忽略警告
warnings.filterwarnings('ignore')

# 设置字体路径
FONT_PATH = r'../simkai.ttf'
font_prop = fm.FontProperties(fname=FONT_PATH)
plt.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False

# 示例数据
MAJORS = ['营销', '金融', '心理学', '新闻学', '电视', '日语', '汉语言', '物理', '机械', '文学', '生物', '计算机', '化学', '数学', '设计', '英语', '会计', '教育', '工程', '医学']
DATAS = [18.7, 21.6, 22.9, 23.5, 24.8, 26.2, 27.1, 27.1, 29.6, 31.3, 32.6, 33.5, 36.3, 40.8, 54.3, 55.8, 70.1, 70.3, 70.7, 100]
MAJORS = [
    '营销', '金融', '心理学', '新闻学', '电视', '日语', '汉语言', '物理', '机械', '文学', 
    '生物', '计算机', '化学', '数学', '设计', '英语', '会计', '教育', '工程', '医学', 
    '法学', '历史学', '地理学', '环境科学', '艺术'
]
DATAS = [
    18.5, 21.8, 23.1, 23.3, 25.0, 26.2, 27.1, 27.1, 29.6, 31.3, 
    32.6, 33.5, 36.3, 40.8, 54.3, 55.8, 70.1, 70.3, 70.7, 100, 
    28.5, 19.4, 22.3, 35.7, 48.9
]
# 常量配置
FIGURE_SIZE = (12, 12)
DPI = 100
AXES_POSITION = [0.22, 0.2, 0.6, 0.6]
THETA_GROUP = np.linspace(0, 1, 21) * np.pi
COLORS = ['#fafbff', '#e3effd']  # 扇形区域颜色
BAR_COLORS = ['#6785f2', '#7171fe']  # 柱状图颜色
DASHED_LINE_COLOR = '#9fa0a0'
WHITE_BOUNDARY_COLOR = 'w'

def setup_polar_plot():
    """设置极坐标图的基本配置"""
    fig, ax = plt.subplots(figsize=FIGURE_SIZE, dpi=DPI)
    ax1 = fig.add_axes(AXES_POSITION, polar=True)
    ax.get_yaxis().set_visible(False)
    ax.get_xaxis().set_visible(False)
    for spine in ax.spines.values():
        spine.set_visible(False)
    return fig, ax1

def draw_sector_regions(ax1):
    """绘制扇形区域"""
    for idx, (start, end) in enumerate(funcy.pairwise(THETA_GROUP)):
        ax1.fill_between([start, end], 1.4, 4, facecolor=COLORS[idx % 2])

def draw_dashed_lines(ax1):
    """绘制虚线"""
    for start, end in funcy.pairwise(THETA_GROUP):
        theta = (start + end) / 2
        ax1.plot([theta, theta], [1.4, 3.5], linestyle='dashed', color=DASHED_LINE_COLOR, linewidth=0.25)

def draw_bars(ax1):
    """绘制柱状图"""
    for idx, (start, end) in enumerate(funcy.pairwise(THETA_GROUP)):
        theta = (start + end) / 2
        ax1.bar(theta, 0.025 * DATAS[idx], width=np.pi / 21, bottom=1.4,
                facecolor=BAR_COLORS[idx % 2], edgecolor='white', linewidth=0.1, alpha=0.95, zorder=9)

def draw_white_boundaries(ax1):
    """绘制扇形区域白色边界"""
    for theta in THETA_GROUP:
        ax1.plot([theta, theta], [1, 4], color=WHITE_BOUNDARY_COLOR, linewidth=0.5)

def annotate_majors_and_datas(ax1):
    """标注专业名称和数据"""
    for idx, (start, end) in enumerate(funcy.pairwise(THETA_GROUP)):
        angle = ((start + end) * 0.5 / np.pi) * 180 - (0 if idx < (len(MAJORS) / 2) else 180)
        # 标注专业名称
        ax1.annotate(MAJORS[idx], xy=[(start + end) / 2, 4.4], va='center', ha='center', zorder=10, rotation=angle, fontsize=10)
        # 标注数据
        ax1.annotate(DATAS[idx], xy=[(start + end) / 2, 3.79], va='center', ha='center', zorder=10, rotation=angle, fontsize=10)

def main():
    """主函数"""
    fig, ax1 = setup_polar_plot()
    draw_sector_regions(ax1)
    draw_dashed_lines(ax1)
    draw_bars(ax1)
    draw_white_boundaries(ax1)
    annotate_majors_and_datas(ax1)
    plt.show()

if __name__ == '__main__':
    main()
```
