# 嵌套环形图

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体  
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号
# 创建画布和子图
fig, ax = plt.subplots(figsize=(8, 8))

# 设置嵌套饼图的参数
size = 0.3  # 内层饼图的宽度

# 示例数据：公司年度销售数据
# 外层：三个地区的销售总额
# 内层：每个地区的三种产品类别的销售数据
vals = np.array([
    [60, 32, 25],  # 地区 A：产品 1、产品 2、产品 3
    [37, 40, 20],  # 地区 B：产品 1、产品 2、产品 3
    [29, 10, 15],  # 地区 C：产品 1、产品 2、产品 3
])

# Pastel 柔和色系
outer_colors = ['#A6CEE3', '#B2DF8A', '#FB9A99']  # 外层颜色：浅蓝、浅绿、浅橙
inner_colors = [
    '#C6DBEF', '#D9F0A3', '#FCAE91',  # 地区 A 的产品颜色
    '#9ECAE1', '#A1D99B', '#FC9272',  # 地区 B 的产品颜色
    '#6BAED6', '#74C476', '#FB6A4A',  # 地区 C 的产品颜色
]

# 绘制外层饼图（地区销售总额）
ax.pie(
    vals.sum(axis=1),  # 计算每个地区的销售总额
    radius=1,  # 外层饼图的半径
    colors=outer_colors,  # 外层颜色
    wedgeprops=dict(width=size, edgecolor='w', linewidth=2),  # 扇区样式
    labels=['地区 A', '地区 B', '地区 C'],  # 地区标签
    startangle=90,  # 起始角度
    textprops={'fontsize': 12, 'color': 'black'},  # 文本样式
)

# 绘制内层饼图（产品类别销售数据）
ax.pie(
    vals.flatten(),  # 将数据展平为一维数组
    radius=1 - size,  # 内层饼图的半径
    colors=inner_colors,  # 内层颜色
    wedgeprops=dict(width=size, edgecolor='w', linewidth=2),  # 扇区样式
    startangle=90,  # 起始角度
    labels=[
        '地区 A - 产品 1', '地区 A - 产品 2', '地区 A - 产品 3',
        '地区 B - 产品 1', '地区 B - 产品 2', '地区 B - 产品 3',
        '地区 C - 产品 1', '地区 C - 产品 2', '地区 C - 产品 3',
    ],  # 产品类别标签
    labeldistance=0.75,  # 标签距离圆心的距离
    textprops={'fontsize': 10, 'color': 'black'},  # 文本样式
)

# 设置标题
ax.set_title("各地区及产品类别年度销售分布", fontsize=14, pad=20)

# 显示图形
plt.tight_layout()
plt.savefig("嵌套环形图.png", dpi=300)
plt.show()
```
