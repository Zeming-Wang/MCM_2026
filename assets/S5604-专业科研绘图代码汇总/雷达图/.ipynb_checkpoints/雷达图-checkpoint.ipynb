## 雷达图

雷达图（Radar Chart），也称为蜘蛛图（Spider Chart）或星图（Star Plot），是一种多维数据可视化图表。它将多个变量的数据点绘制在从中心向外辐射的轴上，形成一个多边形。每个轴代表一个变量，数据点的值决定了多边形在该轴上的位置。雷达图常用于比较多个实体的多个属性或维度。

使用场景  
性能评估：  
比较不同实体（如员工、产品、团队）在多个维度上的表现。  
示例：评估员工在沟通、技术、领导力等方面的能力。  

产品分析：  
比较不同产品在多个属性上的表现。  
示例：比较手机在屏幕、电池、相机、性能等方面的评分。  
  
体育分析：  
比较运动员在不同技术指标上的表现。   
示例：比较足球运动员在速度、射门、传球、防守等方面的能力。  
   
市场研究：  
比较不同品牌在多个消费者评价维度上的表现。    
示例：比较汽车品牌在价格、安全性、舒适性、油耗等方面的评分。  

假设我们需要比较两个产品（Product A 和 Product B）在五个维度上的表现：  
设计（Design）  
性能（Performance）  
电池（Battery）  
相机（Camera）  
价格（Price）  

```python
import numpy as np
import matplotlib.pyplot as plt
# 数据
categories = ['Design', 'Performance', 'Battery', 'Camera', 'Price']
product_a = [4, 3, 5, 4, 2]  # Product A 的评分
product_b = [3, 4, 3, 5, 3]  # Product B 的评分

# 将数据闭合（雷达图需要闭合数据）
product_a += [product_a[0]]    # 添加第一个值到末尾
product_b += [product_b[0]]    # 添加第一个值到末尾

# 设置角度（雷达图是圆形的，需要将类别均匀分布在圆周上）
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()

# 创建画布
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'polar': True})
angles = np.append(angles, 0) # 将雷达图首尾链接起来
categories += categories[:1] # 将雷达图首尾链接起来

# 绘制 Product A
ax.plot(angles, product_a, 'o-', linewidth=2, label='Product A', color='blue')
ax.fill(angles, product_a, color='blue', alpha=0.25)

# 绘制 Product B
ax.plot(angles, product_b, 'o-', linewidth=2, label='Product B', color='green')
ax.fill(angles, product_b, color='green', alpha=0.25)

# print(len(angles))
# print(len(categories))
# 设置角度刻度标签
ax.set_xticks(angles)
ax.set_xticklabels(categories[:], fontsize=12)

# 设置 y 轴标签
ax.set_yticklabels([])

# 添加图例
plt.legend(loc='upper right', bbox_to_anchor=(1.1, 1.1), fontsize=12)

# 添加标题
plt.title('Product Comparison - Radar Chart', fontsize=16, pad=20)

# 显示图表
plt.show()
```
