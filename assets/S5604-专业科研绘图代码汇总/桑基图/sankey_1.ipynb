```python
import numpy as np
import pandas as pd
import json
import requests
from generate_sankey import SanKey
```

```python
# 处理数据为三列，第一列source，字符串；第二列target，字符串；第三列value，数值。
# 所使用数据来源 https://cdn.jsdelivr.net/gh/apache/echarts-website@asf-site/examples/data/asset/data/energy.json
# 读取数据
with open('energy.json', 'r') as f:
    content = f.read()
print(content)
```

```python
content_json = json.loads(content)
# print(content_json['links'])
links = content_json['links']
dataPd = pd.concat([pd.DataFrame({key: [item] for key, item in data.items()}) for data in links], ignore_index=True)
print(dataPd.head(10))
```

```python
sankey_obj = SanKey(pandasdf=dataPd, title="Energy")
sankey_obj()
sankey_obj.save_html(filename="sankey_20240921.html")
```
