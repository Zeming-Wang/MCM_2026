```python
from pyecharts.charts import Bar, Pie, Line
from pyecharts import options as opts
from pyecharts.commons.utils import JsCode
from pyecharts.globals import ThemeType

# 示例数据
x_data = ["衬衫", "羊毛衫", "雪纺衫", "裤子", "高跟鞋", "袜子"]
y_data1 = [5, 20, 36, 10, 75, 90]
y_data2 = [15, 25, 16, 55, 48, 8]
pie_data = [("衬衫", 5), ("羊毛衫", 20), ("雪纺衫", 36), ("裤子", 10), ("高跟鞋", 75), ("袜子", 90)]

# 创建第一个图表 - 柱状图
bar = (
    Bar(init_opts=opts.InitOpts(theme=ThemeType.LIGHT, width="45%", height="400px"))
    .add_xaxis(x_data)
    .add_yaxis("商家A", y_data1)
    .add_yaxis("商家B", y_data2)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="柱状图示例"),
        tooltip_opts=opts.TooltipOpts(trigger="axis"),
        legend_opts=opts.LegendOpts(pos_right="5%"),
    )
)

# 创建第二个图表 - 饼图
pie = (
    Pie(init_opts=opts.InitOpts(theme=ThemeType.LIGHT, width="45%", height="400px"))
    .add(
        "",
        pie_data,
        radius=["30%", "75%"],
        center=["50%", "50%"],
        rosetype="radius",
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="饼图示例"),
        legend_opts=opts.LegendOpts(orient="vertical", pos_top="15%", pos_left="2%"),
    )
    .set_series_opts(
        label_opts=opts.LabelOpts(formatter="{b}: {c} ({d}%)")
    )
)

# 创建第三个图表 - 折线图
line = (
    Line(init_opts=opts.InitOpts(theme=ThemeType.LIGHT, width="90%", height="400px"))
    .add_xaxis(x_data)
    .add_yaxis("商家A", y_data1, is_smooth=True)
    .add_yaxis("商家B", y_data2, is_smooth=True)
    .set_global_opts(
        title_opts=opts.TitleOpts(title="折线图示例", pos_top="48%"),
        tooltip_opts=opts.TooltipOpts(trigger="axis"),
        legend_opts=opts.LegendOpts(pos_top="48%"),
    )
)

# 使用Page类将多个图表组合到一个HTML文件中
from pyecharts.charts import Page

page = Page(layout=Page.SimplePageLayout)  # 简单布局，图表垂直排列
page.add(bar, pie, line)

# 渲染所有图表到一个HTML文件
page.render("multiple_charts.html")
```
