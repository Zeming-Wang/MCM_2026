```python
import pyecharts.options as opts
from pyecharts.charts import Bar3D
import random
import math

def generate_time_labels():
    """生成24小时时间标签"""
    return [f"{i}a" if i < 12 else f"{i-12}p" if i > 12 else "12p" for i in range(24)]

def generate_day_labels():
    """生成星期标签"""
    return ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]

def simulate_activity_pattern(hour, day_index):
    """
    模拟基于时间和星期的活动模式
    参数:
        hour: 当前小时 (0-23)
        day_index: 星期索引 (0-6)
    返回: 模拟的活动值
    """
    # 基础波动 (日夜差异)
    base = 5 * math.sin(hour/24 * math.pi)  
    
    # 工作日/周末差异
    if day_index < 5:  # 工作日
        base += 3 * math.sin((hour-9)/8 * math.pi) + random.randint(3, 8)  # 工作时间高峰
    else:  # 周末
        base += 2 * math.sin((hour-11)/6 * math.pi)  # 周末晚起
    
    # 随机波动
    noise = random.gauss(0, 1.5)
    
    # 确保非负并取整
    return max(0, int(base + noise))

def generate_3d_data():
    """生成3D柱状图模拟数据"""
    time_labels = generate_time_labels()
    day_labels = generate_day_labels()
    
    activity_data = []
    for day_idx, day in enumerate(day_labels):
        for hour_idx, hour in enumerate(time_labels):
            value = simulate_activity_pattern(hour_idx, day_idx)
            # 格式: [x轴索引, y轴索引, 值]
            activity_data.append([day_idx, hour_idx, value])
    
    return activity_data, time_labels, day_labels

def create_3d_bar_chart():
    """创建3D柱状图"""
    chart_data, time_labels, day_labels = generate_3d_data()
    
    # 重新排序数据格式 (pyecharts需要[y, x, value])
    formatted_data = [[d[1], d[0], d[2]] for d in chart_data]
    
    # 创建图表
    activity_chart = (
        Bar3D()
        .add(
            series_name="活动强度",
            data=formatted_data,
            xaxis3d_opts=opts.Axis3DOpts(type_="category", data=time_labels, name="时间"),
            yaxis3d_opts=opts.Axis3DOpts(type_="category", data=day_labels, name="星期"),
            zaxis3d_opts=opts.Axis3DOpts(type_="value", name="活动量"),
        )
        .set_global_opts(
            visualmap_opts=opts.VisualMapOpts(
                max_=20,
                range_color=[
                    "#313695",
                    "#4575b4",
                    "#74add1",
                    "#abd9e9",
                    "#e0f3f8",
                    "#ffffbf",
                    "#fee090",
                    "#fdae61",
                    "#f46d43",
                    "#d73027",
                    "#a50026",
                ],
                pos_top="15%",
                pos_left="90%",
            ),
            title_opts=opts.TitleOpts(title="每周活动模式模拟", subtitle="基于时间和星期的活动强度"),
        )
    )
    return activity_chart

def save_and_render_chart(chart, filename):
    """保存并渲染图表"""
    chart.render(filename)
    print(f"图表已保存为 {filename}")

if __name__ == "__main__":
    # 设置随机种子确保可重复性
    random.seed(42)
    
    # 生成并保存图表
    visualization = create_3d_bar_chart()
    save_and_render_chart(visualization, "activity_simulation_3d.html")
```
