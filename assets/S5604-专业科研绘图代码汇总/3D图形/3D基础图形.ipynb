```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D

import matplotlib.font_manager as fm
# 设置字体路径
font_path = r'../simkai.ttf'  # Windows示例
 
# 加载字体并设置
font_prop = fm.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False

# 创建数据（减少网格点数量）
x = np.linspace(-5, 5, 15)  # 从100减少到30
y = np.linspace(-5, 5, 15)
X, Y = np.meshgrid(x, y)

# 生成更简单的凹凸表面
Z = np.sin(np.sqrt(X**2 + Y**2))  # 基础波形
Z += 0.3 * np.sin(2*X)           # 减弱X方向波动
Z += 0.3 * np.cos(3*Y)           # 减弱Y方向波动

# 创建图形
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

# 绘制表面图（增加网格跨度）
surf = ax.plot_surface(
    X, Y, Z,
    cmap=cm.viridis,
    edgecolor='k',  # 添加黑色网格线
    linewidth=0.3,
    rstride=3,      # 增大行跨度
    cstride=3       # 增大列跨度
)

# 添加颜色条
cbar = fig.colorbar(surf, shrink=0.6, aspect=8)
cbar.set_label('高度值', rotation=270, labelpad=15)

# 设置视角
ax.view_init(elev=40, azim=30)

# 添加标签
ax.set_xlabel('X轴', labelpad=10)
ax.set_ylabel('Y轴', labelpad=10)
ax.set_zlabel('Z轴', labelpad=10)
ax.set_title('简化版三维表面', pad=15)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.font_manager as fm
# 设置字体路径
font_path = r'../simkai.ttf'  # Windows示例
 
# 加载字体并设置
font_prop = fm.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False
# 创建数据
x = np.linspace(-5, 5, 20)
y = np.linspace(-5, 5, 20)
X, Y = np.meshgrid(x, y)

# 生成凹凸不平的Z值（多个正弦波叠加）
Z = np.sin(np.sqrt(X**2 + Y**2))    # 基础波形
Z += 0.5 * np.sin(2*X)              # 添加X方向波动
Z += 0.5 * np.cos(3*Y)              # 添加Y方向波动
Z += 0.3 * np.random.randn(*Z.shape) # 添加随机噪声

# 创建图形
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 绘制表面图（颜色映射）
surf = ax.plot_surface(
    X, Y, Z,
    cmap=cm.viridis,          # 使用Viridis颜色映射
    linewidth=0,
    antialiased=True,
    rstride=2,                # 网格行跨度（控制密度）
    cstride=2                 # 网格列跨度
)

# 添加颜色条
cbar = fig.colorbar(surf, shrink=0.5, aspect=10)
cbar.set_label('Z值高度', rotation=270, labelpad=15)

# 设置视角
ax.view_init(elev=30, azim=45)  # 仰角30度，方位角45度

# 添加标签和标题
ax.set_xlabel('X轴')
ax.set_ylabel('Y轴')
ax.set_zlabel('Z轴')
ax.set_title('三维凹凸平面（颜色随高度变化）', pad=20)

# 显示图形
plt.tight_layout()
plt.show()
```
