```python
from pyecharts import options as opts
from pyecharts.charts import Map
from pyecharts.globals import ThemeType

# 准备数据
data = [
    ("Norway", 7.82),
    ("Netherlands", 8.50),
    ("Portugal", 7.38),
    ("Germany", 8.99),
    ("France", 3.98),
    ("Sweden", 0.32),
    ("Finland", 0.02),
    ("China", 10.00),
    ("Australia", 4.00),
    ("United States", 8.00),
    ("Russia", 6.00),
    ("India", 3.00),
    ("Brazil", 3.00),
    ("Lybia", 7.00),
    ("United Kingdom", 3.00),
    ("Algeria", 2.00),
    ("Argentina", 4.00),
    ("Canada", 5.00),
    ("Egypt", 3.00),
    ("Sudan", 4.00),
    ("Pakistan", 3.00),
    ("Tanzania", 1.00),
]
# 创建紫色科技主题地图
world_map = (
    Map(init_opts=opts.InitOpts(
        width="1200px", 
        height="700px",
        bg_color="#110a1d",  # 深紫色背景
        theme=ThemeType.DARK  # 使用暗黑主题作为基础
    ))
    .add(
        series_name="占比(%)",
        data_pair=data,
        maptype="world",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(
            is_show=False,
            color="#d1c4e9"  # 浅紫色文字
        ),
        itemstyle_opts=opts.ItemStyleOpts(
            border_width=0.8,
            border_color="rgba(179,136,255,0.5)",  # 紫色边框
            area_color="rgba(20,10,40,0.8)"  # 深紫色区域
        ),
        zoom=1.2,
        emphasis_itemstyle_opts=opts.ItemStyleOpts(
            area_color="rgba(101,31,255,0.4)"  # 高亮紫色
        ),
        emphasis_label_opts=opts.LabelOpts(
            is_show=True,
            color="#b388ff"
        ),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter="{b}: {c}%",
            background_color="rgba(17,10,29,0.9)",  # 深紫色背景
            border_color="#7c4dff",  # 紫色边框
            textstyle_opts=opts.TextStyleOpts(color="#d1c4e9")
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="世界各国占比热力图 - 紫色科技风格",
            subtitle="数据来源：示例数据",
            pos_left="center",
            title_textstyle_opts=opts.TextStyleOpts(
                font_size=24,
                color="#b388ff",  # 紫色标题
                font_weight="bold"
            ),
            subtitle_textstyle_opts=opts.TextStyleOpts(
                color="#7c4dff",  # 中等紫色副标题
                font_size=14
            )
        ),
        visualmap_opts=opts.VisualMapOpts(
            max_=10,
            min_=0,
            is_piecewise=False,
            range_color=["#b388ff", "#7c4dff", "#651fff"],  # 浅紫-中紫-深紫渐变
            orient="horizontal",
            pos_left="center",
            pos_bottom="10%",
            textstyle_opts=opts.TextStyleOpts(color="#d1c4e9"),
            background_color="rgba(17,10,29,0.7)",
            border_color="#b388ff",
            border_width=1
        ),
        legend_opts=opts.LegendOpts(
            is_show=False
        )
    )
)

# 渲染地图
world_map.render("紫色科技风格_world_map.html")
```
