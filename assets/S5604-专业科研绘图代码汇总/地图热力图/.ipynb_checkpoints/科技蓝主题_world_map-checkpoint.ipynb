```python
from pyecharts import options as opts
from pyecharts.charts import Map
from pyecharts.globals import ThemeType

# 准备数据
data = [
    ("Norway", 7.82),
    ("Netherlands", 8.50),
    ("Portugal", 7.38),
    ("Germany", 8.99),
    ("France", 3.98),
    ("Sweden", 0.32),
    ("Finland", 0.02),
    ("China", 10.00),
    ("Australia", 4.00),
    ("United States", 8.00),
    ("Russia", 6.00),
    ("India", 3.00),
    ("Brazil", 3.00),
    ("Lybia", 7.00),
    ("United Kingdom", 3.00),
    ("Algeria", 2.00),
    ("Argentina", 4.00),
    ("Canada", 5.00),
    ("Egypt", 3.00),
    ("Sudan", 4.00),
    ("Pakistan", 3.00),
    ("Tanzania", 1.00),
]

# 创建科技蓝主题地图
world_map = (
    Map(init_opts=opts.InitOpts(
        width="1300px", 
        height="750px",
        bg_color="#0a1128",  # 深蓝色太空背景
        theme=ThemeType.DARK,
        chart_id="tech_blue_map"
    ))
    .add(
        series_name="数据占比(%)",
        data_pair=data,
        maptype="world",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(
            is_show=False,
            color="#a0e7ff"
        ),
        itemstyle_opts=opts.ItemStyleOpts(
            border_width=1,
            border_color="rgba(0, 180, 255, 0.6)",
            area_color="rgba(10, 30, 60, 0.8)",
            border_type="solid"
        ),
        zoom=1.2,
        emphasis_itemstyle_opts=opts.ItemStyleOpts(
            area_color="rgba(0, 150, 255, 0.5)",
            border_width=1.5,
            border_color="rgba(0, 240, 255, 0.9)"
        ),
        emphasis_label_opts=opts.LabelOpts(
            is_show=True,
            color="#00f4ff",
            font_size=12,
            font_weight="bold"
        ),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter="{b}: {c}%",
            background_color="rgba(10, 17, 40, 0.95)",
            border_color="#00acff",
            border_width=1,
            textstyle_opts=opts.TextStyleOpts(
                color="#c3e7ff",
                font_size=14
            )
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="全球数据分布热力图",
            subtitle="科技蓝主题 | 数据可视化",
            pos_left="center",
            pos_top="0%",
            title_textstyle_opts=opts.TextStyleOpts(
                font_size=28,
                color="#00f4ff",
                font_weight="bolder",
                font_family="Arial"
            ),
            subtitle_textstyle_opts=opts.TextStyleOpts(
                color="#3d7bff",
                font_size=16,
                font_family="Arial"
            )
        ),
        visualmap_opts=opts.VisualMapOpts(
            max_=10,
            min_=0,
            is_piecewise=False,
            range_color=["#50a3ba", "#eac736", "#d94e5d"],
            orient="horizontal",
            pos_left="15%",
            pos_right="15%",
            pos_bottom="8%",
            textstyle_opts=opts.TextStyleOpts(
                color="#a0e7ff",
                font_size=12
            ),
            background_color="rgba(10, 17, 40, 0.7)",
            border_color="#00acff",
            border_width=1
        ),
        legend_opts=opts.LegendOpts(
            is_show=False
        )
    )
)

# 渲染地图
world_map.render("科技蓝主题_world_map.html")
```
