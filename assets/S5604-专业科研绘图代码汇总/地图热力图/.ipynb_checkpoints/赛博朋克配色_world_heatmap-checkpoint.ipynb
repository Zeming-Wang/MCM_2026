```python
from pyecharts import options as opts
from pyecharts.charts import Map
from pyecharts.globals import ThemeType

# 准备数据
data = [
    ("Norway", 7.82),
    ("Netherlands", 8.50),
    ("Portugal", 7.38),
    ("Germany", 8.99),
    ("France", 3.98),
    ("Sweden", 0.32),
    ("Finland", 0.02),
    ("China", 10.00),
    ("Australia", 4.00),
    ("United States", 8.00),
    ("Russia", 6.00),
    ("India", 3.00),
    ("Brazil", 3.00),
    ("Lybia", 7.00),
    ("United Kingdom", 3.00),
    ("Algeria", 2.00),
    ("Argentina", 4.00),
    ("Canada", 5.00),
    ("Egypt", 3.00),
    ("Sudan", 4.00),
    ("Pakistan", 3.00),
    ("Tanzania", 1.00),
]

# 创建地图（使用DARK主题作为基础）
world_map = (
    Map(init_opts=opts.InitOpts(
        width="1000px", 
        height="600px",
        bg_color="#1a1a2e",  # 赛博朋克风格背景色
        theme=ThemeType.DARK  # 使用暗黑主题作为基础
    ))
    .add(
        series_name="占比(%)",
        data_pair=data,
        maptype="world",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(
            is_show=False,
            color="#d1f7ff"  # 赛博朋克风格文字颜色
        ),
        itemstyle_opts=opts.ItemStyleOpts(
            border_width=0.5,
            border_color="rgba(255,255,255,0.3)",
            area_color="rgba(30,30,60,1)"  # 深色区域
        ),
        zoom=1.2,
        emphasis_itemstyle_opts=opts.ItemStyleOpts(
            area_color="rgba(255,42,109,0.3)"  # 赛博朋克粉高亮
        ),
        emphasis_label_opts=opts.LabelOpts(
            is_show=True,
            color="#05d9e8"  # 赛博朋克青蓝色
        ),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter="{b}: {c}%",
            background_color="rgba(26,26,46,0.9)",  # 赛博朋克背景色
            border_color="#ff2a6d",  # 赛博朋克粉色边框
            textstyle_opts=opts.TextStyleOpts(color="#d1f7ff")
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="世界各国占比热力图 - 赛博朋克风格",
            subtitle="数据来源：示例数据",
            pos_left="center",
            title_textstyle_opts=opts.TextStyleOpts(
                font_size=20,
                color="#05d9e8"  # 青蓝色标题
            ),
            subtitle_textstyle_opts=opts.TextStyleOpts(
                color="#ff2a6d"  # 粉色副标题
            )
        ),
        visualmap_opts=opts.VisualMapOpts(
            max_=25,
            min_=0,
            is_piecewise=False,
            range_color=["#ff2a6d", "#05d9e8", "#005678"],  # 赛博朋克配色
            orient="horizontal",
            pos_left="center",
            pos_bottom="10%",
            textstyle_opts=opts.TextStyleOpts(color="#d1f7ff"),
            background_color="rgba(26,26,46,0.7)",
            border_color="#ff2a6d"
        ),
        legend_opts=opts.LegendOpts(is_show=False)
    )
)

# 渲染地图
world_map.render("赛博朋克配色_world_heatmap.html")
```
