```python
from pyecharts import options as opts
from pyecharts.charts import Map

# 准备数据（使用英文国家名称）
data = [
    ("Norway", 7.82),
    ("Netherlands", 8.50),
    ("Portugal", 7.38),
    ("Germany", 8.99),
    ("France", 3.98),
    ("Sweden", 0.32),
    ("Finland", 0.02),
    ("China", 10.00),
    ("Australia", 4.00),
    ("United States", 8.00),
    ("Russia", 6.00),
    ("India", 3.00),
    ("Brazil", 3.00),
    ("Lybia", 7.00),
    ("United Kingdom", 3.00),
    ("Algeria", 2.00),
    ("Argentina", 4.00),
    ("Canada", 5.00),
    ("Egypt", 3.00),
    ("Sudan", 4.00),
    ("Pakistan", 3.00),
    ("Tanzania", 1.00),
]

# 创建地图
world_map = (
    Map(init_opts=opts.InitOpts(width="1000px", height="600px", bg_color="#FFFFFF"))
    .add(
        series_name="占比(%)",
        data_pair=data,
        maptype="world",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(is_show=False),
        itemstyle_opts=opts.ItemStyleOpts(
            border_width=0.5,
            border_color="rgba(0,0,0,0.2)"
        ),
        zoom=1.2,
        emphasis_label_opts=opts.LabelOpts(is_show=True),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter="{b}: {c}%"
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="世界各国占比热力图",
            subtitle="数据来源：示例数据",
            pos_left="center",
            title_textstyle_opts=opts.TextStyleOpts(font_size=20)
        ),
        visualmap_opts=opts.VisualMapOpts(
            max_=35,
            min_=0,
            is_piecewise=False,
            range_color=["#50a3ba", "#eac736", "#d94e5d"],
            orient="horizontal",
            pos_left="center",
            pos_bottom="10%"
        ),
        legend_opts=opts.LegendOpts(is_show=False)
    )
)

# 渲染地图
world_map.render("基础风格_world_heatmap.html")
```
