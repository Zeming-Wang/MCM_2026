```python
from pyecharts import options as opts
from pyecharts.charts import Map
from pyecharts.globals import ThemeType

# 数据准备
data = [
    ("Norway", 7.82),
    ("Netherlands", 8.50),
    ("Portugal", 7.38),
    ("Germany", 8.99),
    ("France", 3.98),
    ("Sweden", 0.32),
    ("Finland", 0.02),
    ("China", 10.00),
    ("Australia", 4.00),
    ("United States", 8.00),
    ("Russia", 6.00),
    ("India", 3.00),
    ("Brazil", 3.00),
    ("Lybia", 7.00),
    ("United Kingdom", 3.00),
    ("Algeria", 2.00),
    ("Argentina", 4.00),
    ("Canada", 5.00),
    ("Egypt", 3.00),
    ("Sudan", 4.00),
    ("Pakistan", 3.00),
    ("Tanzania", 1.00),
]

world_map = (
    Map(init_opts=opts.InitOpts(
        width="1000px",
        height="600px",
        theme=ThemeType.DARK,
        bg_color="#0a0a0a"
    ))
    .add(
        series_name="占比(%)",
        data_pair=data,
        maptype="world",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(
            is_show=False,
            color="#fff"  # 白色标签
        ),
        itemstyle_opts=opts.ItemStyleOpts(
            border_width=0.5,
            border_color="rgba(255,255,255,0.3)",  # 白色边框
            area_color="rgba(20,20,20,1)"  # 深色区域
        ),
        zoom=1.2,
        emphasis_itemstyle_opts=opts.ItemStyleOpts(
            area_color="rgba(255,255,255,0.1)"  # 高亮颜色
        ),
        emphasis_label_opts=opts.LabelOpts(
            is_show=True,
            color="#fff"
        ),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter="{b}: {c}%",
            background_color="rgba(0,0,0,0.7)",  # 黑色半透明
            border_color="#333",
            textstyle_opts=opts.TextStyleOpts(color="#fff")
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="世界各国占比热力图",
            subtitle="数据来源：示例数据",
            pos_left="center",
            title_textstyle_opts=opts.TextStyleOpts(
                font_size=20,
                color="#fff"  # 白色标题
            ),
            subtitle_textstyle_opts=opts.TextStyleOpts(
                color="#ccc"  # 浅灰色副标题
            )
        ),
        visualmap_opts=opts.VisualMapOpts(
            max_=10,
            min_=0,
            is_piecewise=False,
            range_color=["#ff2a6d", "#05d9e8", "#005678"],  # 蓝-金-红渐变
            orient="horizontal",
            pos_left="center",
            pos_bottom="10%",
            textstyle_opts=opts.TextStyleOpts(color="#fff"),  # 白色文字
            background_color="rgba(0,0,0,0.5)",  # 半透明黑色背景
            border_color="#333"
        ),
        legend_opts=opts.LegendOpts(is_show=False)
    )
)

# 渲染地图
world_map.render("黑色主题_world_heatmap-theme.html")
```
