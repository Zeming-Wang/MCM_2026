```python
# 导入相关的包
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots()

# 修改后的水果种类和对应数量
fruits = ['banana', 'grape', 'kiwi', 'strawberry']
counts = [75, 50, 90, 60]
# 每种水果对应的标签
bar_labels = ['yellow', 'purple', 'green', 'red']
# 每种水果对应的颜色
bar_colors = ['yellow', 'purple', 'green', 'red']

# 生成条形图
ax.bar(fruits, counts, label=bar_labels, color=bar_colors)

# 设置标签和标题
ax.set_ylabel('Fruit supply')
ax.set_title('New Fruit supply by kind and color')
ax.legend(title='Fruit color')

# 显示图形
plt.show()
```

```python
# 定义新的数据场景
car_models = ('Sedan', 'SUV', 'Truck')
yearly_sales = {
    '2020': np.array([150, 200, 180]),
    '2021': np.array([160, 220, 190]),
    '2022': np.array([170, 210, 200]),
}
width = 0.6  # 条形宽度

# 创建绘图
fig, ax = plt.subplots()
bottom = np.zeros(3)

# 绘制不同年份的汽车销量
for year, sales in yearly_sales.items():
    p = ax.bar(car_models, sales, width, label=year, bottom=bottom)
    bottom += sales  # 堆叠条形图

    # 在条形图上显示销量数据
    ax.bar_label(p, label_type='center')

# 设置标题和图例
ax.set_title('Car sales by model and year')
ax.legend(title='Year')

# 显示图形
plt.show()
```

```python
# 定义数据场景
fruits = ("Apple", "Banana", "Orange")
fruit_means = {
    'Weight (g)': (182, 118, 131),
    'Vitamin C (mg)': (8.4, 10.3, 53.2),
    'Water Content (%)': (85.6, 74.9, 86.8),
}

x = np.arange(len(fruits))  # 标签位置
width = 0.25  # 条形的宽度
multiplier = 0

# 创建绘图
fig, ax = plt.subplots(layout='constrained')

# 绘制水果的不同属性条形图
for attribute, measurement in fruit_means.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, measurement, width, label=attribute)
    ax.bar_label(rects, padding=3)
    multiplier += 1

# 添加标题、标签、图例
ax.set_ylabel('Measurements')
ax.set_title('Fruit attributes comparison')
ax.set_xticks(x + width, fruits)
ax.legend(loc='upper left', ncols=3)
ax.set_ylim(0, 250)

plt.show()
```

```python
# 固定随机数种子
np.random.seed(19680801)

fig, ax = plt.subplots()

# 新的数据场景：不同城市的年平均降雨量
cities = ('New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix')
y_pos = np.arange(len(cities))
rainfall = 800 + 500 * np.random.rand(len(cities))  # 随机生成降雨量数据
error = 50 * np.random.rand(len(cities))  # 随机生成误差范围

# 创建水平条形图
ax.barh(y_pos, rainfall, xerr=error, align='center')
ax.set_yticks(y_pos, labels=cities)
ax.invert_yaxis()  # 让城市标签从上到下排列
ax.set_xlabel('Average Rainfall (mm)')
ax.set_title('Annual Average Rainfall in Different Cities')

plt.show()
```
