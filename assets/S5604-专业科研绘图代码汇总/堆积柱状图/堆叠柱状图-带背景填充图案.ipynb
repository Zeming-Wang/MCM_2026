```python
import matplotlib.pyplot as plt
import numpy as np

# 数据准备
categories = ['A', 'B', 'C', 'D']
series1 = [20, 35, 30, 35]
series2 = [25, 32, 34, 20]
series3 = [12, 15, 18, 25]

# 创建图形
fig, ax = plt.subplots(figsize=(10, 6))

# 定义填充图案
patterns = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']

# 绘制堆叠柱状图（带填充）
bars1 = ax.bar(categories, series1, label='Series 1', 
               hatch=patterns[0], edgecolor='black', alpha=0.8)
bars2 = ax.bar(categories, series2, bottom=series1, label='Series 2', 
               hatch=patterns[1], edgecolor='black', alpha=0.8)
bars3 = ax.bar(categories, series3, bottom=np.array(series1)+np.array(series2), 
               label='Series 3', hatch=patterns[2], edgecolor='black', alpha=0.8)

# 添加数据标签
def add_labels(bars, bottoms=None):
    for i, bar in enumerate(bars):
        height = bar.get_height()
        y = bottoms[i] + height/2 if bottoms is not None else height/2
        ax.text(bar.get_x() + bar.get_width()/2, y, f'{height}', 
                ha='center', va='center', fontsize=10)

add_labels(bars1)
add_labels(bars2, bottoms=series1)
add_labels(bars3, bottoms=np.array(series1)+np.array(series2))

# 美化图形
ax.set_title('Stacked Bar Chart with Patterns', pad=20, fontsize=14)
ax.set_xlabel('Categories', labelpad=10)
ax.set_ylabel('Values', labelpad=10)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(axis='y', linestyle=':', alpha=0.4)

plt.tight_layout()
plt.savefig('堆叠柱状图-带背景填充图案.png', dpi=400, bbox_inches='tight')
plt.show()
```
