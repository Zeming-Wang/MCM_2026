# 基础堆叠柱状图

```python
import matplotlib.pyplot as plt
import numpy as np

# 示例数据
categories = ['A', 'B', 'C', 'D']
series1 = [20, 35, 30, 35]
series2 = [25, 32, 34, 20]
series3 = [12, 15, 18, 25]

# 创建图形
fig, ax = plt.subplots(figsize=(8, 6))

# 绘制堆叠柱状图
ax.bar(categories, series1, label='Series 1')
ax.bar(categories, series2, bottom=series1, label='Series 2')
ax.bar(categories, series3, bottom=np.array(series1)+np.array(series2), label='Series 3')

# 添加标签和标题
ax.set_xlabel('Categories')
ax.set_ylabel('Values')
ax.set_title('Stacked Bar Chart')
ax.legend()

plt.tight_layout()
plt.savefig('基础堆叠柱状图.png', dpi=400, bbox_inches='tight')
plt.show()
```
