```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 生成随机数据
np.random.seed(42)
data1 = np.random.normal(0, 1, 200)  # 数据集1，均值为0，标准差为1
data2 = np.random.normal(5, 2, 200)  # 数据集2，均值为5，标准差为2
data3 = np.random.normal(10, 1.5, 200)  # 数据集3，均值为10，标准差为1.5

# 将数据组合成一个列表
dataset = [data1, data2, data3]
```

| 参数         | 作用                                                                 |
|--------------|----------------------------------------------------------------------|
| dataset      | 输入数据，可以是数组或向量序列。                                       |
| positions    | 小提琴图的位置，默认为 [1, 2, ..., n]。                                |
| vert         | 是否垂直绘制小提琴图（已弃用，推荐使用 orientation）。                  |
| orientation  | 方向，'vertical'（默认）或 'horizontal'。                              |
| widths       | 小提琴图的最大宽度，默认为 0.5。                                       |
| showmeans    | 是否显示均值线，默认为 False。                                         |
| showextrema  | 是否显示极值线（最小值和最大值），默认为 True。                         |
| showmedians  | 是否显示中位数线，默认为 False。                                       |
| quantiles    | 显示指定分位数线，默认为 None。                                        |
| points       | 核密度估计的评估点数，默认为 100。                                     |
| bw_method    | 核密度估计的带宽计算方法，默认为 'scott'。                              |
| side         | 小提琴图的显示方向，'both'（默认）、'low' 或 'high'。                   |
| data         | 如果提供，参数可以接受字符串作为数据的键。                              |

### 绘制基础小提琴图

```python
# 创建画布
plt.figure(figsize=(10, 6))

# 绘制小提琴图
plt.violinplot(dataset)

# 添加标题和标签
plt.title('Basic Violin Plot')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

### 自定义位置和宽度

```python
# 创建画布
plt.figure(figsize=(10, 6))

# 自定义位置和宽度
positions = [1, 2, 4]  # 小提琴图的位置
widths = [0.3, 0.5, 0.7]  # 小提琴图的宽度

# 绘制小提琴图
plt.violinplot(dataset, positions=positions, widths=widths)

# 添加标题和标签
plt.title('Custom Positions and Widths')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

### 显示均值、中位数和分位数

```python
# 创建画布
plt.figure(figsize=(10, 6))

# 绘制小提琴图，显示均值、中位数和分位数
plt.violinplot(
    dataset,
    showmeans=True,  # 显示均值线
    showmedians=True,  # 显示中位数线
    quantiles=[[0.25, 0.75], [0.25, 0.75], [0.25, 0.75]],  # 显示25%和75%分位数
)

# 添加标题和标签
plt.title('Violin Plot with Means, Medians, and Quantiles')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

### 调整带宽和显示单侧

```python
# 创建画布
plt.figure(figsize=(10, 6))

# 绘制小提琴图，调整带宽并显示单侧
plt.violinplot(
    dataset,
    bw_method=0.2,  # 调整带宽
    side='high',  # 只显示上方
)

# 添加标题和标签
plt.title('Violin Plot with Adjusted Bandwidth and Single Side')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

### 设置填充颜色

```python
# 生成随机数据
np.random.seed(42)
data1 = np.random.normal(0, 1, 200)  # 数据集1，均值为0，标准差为1
data2 = np.random.normal(5, 2, 200)  # 数据集2，均值为5，标准差为2
data3 = np.random.normal(10, 1.5, 200)  # 数据集3，均值为10，标准差为1.5

# 将数据组合成一个列表
dataset = [data1, data2, data3]

# 创建画布
plt.figure(figsize=(10, 6))

# 绘制小提琴图
violins = plt.violinplot(
    dataset,
    showmeans=True,  # 显示均值线
    showmedians=True,  # 显示中位数线
    quantiles=[[0.25, 0.75], [0.25, 0.75], [0.25, 0.75]],  # 为每个数据集指定分位数
)

# 设置填充颜色
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # 自定义颜色列表
for body, color in zip(violins['bodies'], colors):
    body.set_facecolor(color)  # 设置填充颜色
    body.set_alpha(0.8)  # 设置透明度

# 设置边界颜色
for partname in ('cbars', 'cmins', 'cmaxes', 'cmeans', 'cmedians', 'cquantiles'):
    part = violins[partname]
    part.set_edgecolor('black')  # 设置边界颜色
    part.set_linewidth(1)  # 设置边界线宽

# 添加标题和标签
plt.title('Violin Plot with Custom Colors', fontsize=16)
plt.xlabel('Category', fontsize=14)
plt.ylabel('Value', fontsize=14)

# 设置x轴刻度标签
plt.xticks([1, 2, 3], ['Category 1', 'Category 2', 'Category 3'], fontsize=12)

# 显示图表
plt.show()
```
