### 小提琴图（Violin Plot）是一种结合了箱线图和核密度估计的图表类型，用于展示数据的分布情况。它能够显示数据的密度分布，尤其是在数据分布不对称或多峰分布时，小提琴图比箱线图提供更多的信息。

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

| 参数      | 作用                                                                 |
|-----------|--------------------------------------------------------------------|
| `data`    | DataFrame 数据集                                                   |
| `x`       | 分类变量                                                            |
| `y`       | 数值变量                                                            |
| `hue`     | 按类别分色                                                          |
| `order`   | 自定义类别顺序                                                      |
| `split`   | `True`：将 `hue` 类别合并显示                                       |
| `palette` | 颜色方案（如 `"Blues"`、`"pastel"`）                                |
| `orient`  | `'v'` 竖向（默认） 或 `'h'` 横向                                    |
| `inner`   | `'box'`（默认）显示箱线图，`'point'` 显示数据点                     |
| `bw_method`   | 可选“scott”, “silverman”                     |
| `density_norm`   | 控制小提琴图的宽度标准化方式。'area': 所有小提琴图的面积相同。'count': 小提琴图的宽度与样本数量成正比。'width': 所有小提琴图的宽度相同。 |
| `bw_adjust`   | 控制绘图的圆滑                     |

## 绘制默认图像

```python
# 生成随机数据
np.random.seed(42)
data1 = np.random.normal(0, 1, 200)
data2 = np.random.normal(5, 2, 200)
data3 = np.random.normal(10, 1.5, 200)

# 将数据放入DataFrame
df = pd.DataFrame({
    'Category': ['A'] * 200 + ['B'] * 200 + ['C'] * 200,
    'Value': np.concatenate([data1, data2, data3])
})
```

```python
# 绘制小提琴图
plt.figure(figsize=(10, 6))
sns.violinplot(x='Category', y='Value', hue='Category', data=df, palette='Set2', legend=False)

# 添加标题和标签
plt.title('Violin Plot Example')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

## 比较不同类别的数据分布

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='Category', y='Value', hue='Category', data=df, palette='Set2', inner='box')
plt.title('Comparing Distributions Across Categories')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

## 分组小提琴图

```python
# 添加分组数据
df['Group'] = ['Group1'] * 100 + ['Group2'] * 100 + ['Group1'] * 100 + ['Group2'] * 100 + ['Group1'] * 100 + ['Group2'] * 100

# 绘制分组小提琴图
plt.figure(figsize=(10, 6))
sns.violinplot(x='Category', y='Value', hue='Group', data=df, palette='Set2', split=True)
plt.title('Grouped Violin Plot')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

### gap参数

```python
# 添加分组数据
df['Group'] = ['Group1'] * 100 + ['Group2'] * 100 + ['Group1'] * 100 + ['Group2'] * 100 + ['Group1'] * 100 + ['Group2'] * 100

# 绘制分组小提琴图
plt.figure(figsize=(10, 6))
sns.violinplot(x='Category', y='Value', hue='Group', data=df, palette='Set2', split=True, gap=.3)
plt.title('Grouped Violin Plot')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

## 调整带宽和宽度

```python
# 绘制小提琴图
plt.figure(figsize=(10, 6))
sns.violinplot(
    x='Category', 
    y='Value', 
    hue='Category', 
    data=df, 
    palette='Set2', 
    bw_method=0.1, 
    density_norm='count',  
    legend=False
)

# 添加标题和标签
plt.title('Violin Plot Example')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```

```python
# 绘制小提琴图
plt.figure(figsize=(10, 6))
sns.violinplot(
    x='Category', 
    y='Value', 
    hue='Category', 
    data=df, 
    palette='Set2', 
    bw_method=0.6,  # 更加圆滑
    density_norm='count',  
    legend=False
)

# 添加标题和标签
plt.title('Violin Plot Example')
plt.xlabel('Category')
plt.ylabel('Value')

# 显示图表
plt.show()
```
