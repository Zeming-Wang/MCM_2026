```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体  
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号
```

```python
plt.style.use('ggplot')

# 保持随机性一致
np.random.seed(42)

# 创建 2x2 子图
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(8, 6))
ax1, ax2, ax3, ax4 = axs.flat

# 子图1：散点图，结合均值和方差不同的分布
x1, y1 = np.random.multivariate_normal([1, 1], [[0.5, 0.2], [0.2, 0.5]], 150).T
x2, y2 = np.random.multivariate_normal([-1, -1], [[0.5, -0.3], [-0.3, 0.5]], 150).T
ax1.scatter(x1, y1, color='b', alpha=0.6, label='Cluster 1')
ax1.scatter(x2, y2, color='r', alpha=0.6, label='Cluster 2')
ax1.set_title("二维正态分布的散点图")
ax1.legend()

# 子图2：多条余弦曲线，使用颜色循环
x = np.linspace(0, 4 * np.pi, 200)
for i, phase_shift in enumerate(np.linspace(0, np.pi, 5)):
    ax2.plot(x, np.cos(x + phase_shift), label=f'Phase {i}')
ax2.set_title("相位偏移的余弦曲线")
ax2.set_xlim([0, 4 * np.pi])
ax2.legend()

# 子图3：堆叠柱状图，使用随机数
categories = ['Cat A', 'Cat B', 'Cat C', 'Cat D', 'Cat E']
data1 = np.random.randint(5, 15, size=5)
data2 = np.random.randint(5, 15, size=5)
ax3.bar(categories, data1, label='Data 1', color='skyblue')
ax3.bar(categories, data2, bottom=data1, label='Data 2', color='lightgreen')
ax3.set_title("堆叠柱状图")
ax3.legend()

# 子图4：随机颜色的矩形填充图
for i in range(6):
    rect = plt.Rectangle(np.random.rand(2) * 2, 0.3, 0.2, color=np.random.rand(3,))
    ax4.add_patch(rect)
ax4.set_xlim(0, 2)
ax4.set_ylim(0, 2)
ax4.set_title("随机矩形填充")
ax4.axis('equal')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('ggplot')

# 固定随机数种子，保证结果可复现
np.random.seed(42)


def generate_beta_hist(ax, alpha, beta, bins=30):
    """在指定轴上绘制贝塔分布直方图"""
    data = np.random.beta(alpha, beta, size=8000)
    ax.hist(data, bins=bins, histtype='step', alpha=0.85, linewidth=1.5, density=True)
    ax.set_ylabel('Density')
    ax.grid(True, linestyle='--', alpha=0.5)


# 创建子图和主图
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 6))

# 使用不同参数生成贝塔分布的直方图
beta_params = [(2, 5), (3, 9), (10, 2), (6, 8)]
titles = ["Alpha=2, Beta=5", "Alpha=3, Beta=9", "Alpha=10, Beta=2", "Alpha=6, Beta=8"]

for ax, params, title in zip(axes.flat, beta_params, titles):
    alpha, beta = params
    generate_beta_hist(ax, alpha, beta)
    ax.set_title(title)

# 设置总标题
fig.suptitle("不同参数下的贝塔分布直方图", fontsize=16, weight='bold')

plt.tight_layout()
plt.subplots_adjust(top=0.9)  # 调整图与标题的距离
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 选择'ggplot'样式
plt.style.use('ggplot')

# 创建 x 轴数据范围
x_vals = np.linspace(0, 10, 100)

# 固定随机数种子，保证结果可复现
np.random.seed(42)

# 创建图形和子图
fig, ax = plt.subplots(figsize=(8, 5))

# 定义不同函数进行绘制
y_funcs = [
    lambda x: np.sin(x) + np.random.normal(0, 0.5, len(x)),
    lambda x: np.cos(x) + 0.3 * x + np.random.normal(0, 0.4, len(x)),
    lambda x: np.sin(2 * x) - 0.2 * x + np.random.normal(0, 0.3, len(x)),
    lambda x: np.cos(0.5 * x) + np.random.normal(0, 0.2, len(x)),
    lambda x: np.sin(0.1 * x) + 0.5 * x + np.random.normal(0, 0.6, len(x)),
]

# 定义不同的线条样式
styles = ['-', '--', '-.', ':', '-']
colors = ['b', 'g', 'r', 'm', 'c']

# 依次绘制不同的函数
for func, style, color in zip(y_funcs, styles, colors):
    ax.plot(x_vals, func(x_vals), linestyle=style, color=color, linewidth=2, alpha=0.8)

# 设置标题和坐标标签
ax.set_title("不同函数的随机波动图", fontsize=16, fontweight='bold')
ax.set_xlabel("X 轴", fontsize=12)
ax.set_ylabel("Y 轴", fontsize=12)

# 显示图例
ax.legend(['sin(x) + noise', 'cos(x) + 0.3x', 'sin(2x) - 0.2x', 
           'cos(0.5x)', 'sin(0.1x) + 0.5x'], loc='upper left')

# 调整布局
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 保持随机数可复现性
np.random.seed(42)

# 使用ggplot样式
plt.style.use('ggplot')

# 创建子图和调整大小
fig, ax = plt.subplots(figsize=(8, 5))

# 定义不同参数的贝塔分布并绘制直方图
beta_params = [(2, 5), (10, 2), (3, 7), (5, 1)]
colors = ['blue', 'green', 'orange', 'red']

# 绘制四个不同参数的贝塔分布直方图
for (a, b), color in zip(beta_params, colors):
    data = np.random.beta(a, b, size=5000)
    ax.hist(data, bins=30, density=True, alpha=0.7, histtype='stepfilled', color=color, label=f'α={a}, β={b}')

# 添加图表标题、图例及轴标签
ax.set_title("不同参数的贝塔分布", fontsize=16, fontweight='bold')
ax.set_xlabel("值", fontsize=12)
ax.set_ylabel("概率密度", fontsize=12)
ax.legend(title='参数组', fontsize=10)

# 显示图表
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 保持随机数一致性，确保数据可复现
np.random.seed(42)

# 定义 x 轴数据范围
x = np.linspace(0, 10)

# 设置新的样式，并结合渐变效果
with plt.style.context('ggplot'):
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # 生成并绘制不同系数的正弦曲线，同时加入噪声
    for factor in range(1, 9):
        y = np.sin(x) + factor * x + np.random.randn(50)
        # 使用不同的颜色和线条样式
        ax.plot(x, y, label=f'Line {factor}', 
                linestyle='-', marker='o', markersize=5, alpha=0.7)
    
    # 设置渐变色填充
    y_fill = np.sin(x) + 4 * x + np.random.randn(50)
    ax.fill_between(x, y_fill - 5, y_fill + 5, color='gray', alpha=0.2, label='Fill Region')
    
    # 配置图表的标题、标签和图例
    ax.set_title("多条正弦曲线与渐变填充", fontsize=16, fontweight='bold')
    ax.set_xlabel("X轴标签", fontsize=12)
    ax.set_ylabel("Y轴标签", fontsize=12)
    ax.legend(loc='upper left', fontsize=10)

# 显示图表
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置字体为黑体  
plt.rcParams['axes.unicode_minus'] = False  # 正确显示负号

# 在0到5秒之间以0.2秒的间隔均匀采样时间
t = np.arange(0., 5., 0.2)

# 生成不同函数的值
y1 = t            # 线性函数
y2 = t ** 2      # 二次函数
y3 = t ** 3      # 三次函数

# 创建一个新的图形，增加标签和标题
plt.figure(figsize=(8, 5))
plt.plot(t, y1, linestyle='-', color='orange', label='y = t')    # 线性函数
plt.plot(t, y2, linestyle='--', marker='o', color='blue', label=r'y = t^2')  # 二次函数
plt.plot(t, y3, linestyle=':', marker='^', color='green', label=r'y = t^3')  # 三次函数

# 添加图例和标题
plt.title('不同函数的图形', fontsize=16)
plt.xlabel('时间 (s)', fontsize=12)
plt.ylabel('函数值', fontsize=12)
plt.legend()

# 显示图形
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import cbook
from mpl_toolkits.axes_grid1 import AxesGrid

def load_demo_image():
    """加载演示用的图像数据，返回数组和坐标范围。"""
    z = cbook.get_sample_data("axes_grid/bivariate_normal.npy")  # 15x15数组
    return z, (-3, 4, -4, 3)

def create_bottom_colorbar(fig):
    """
    创建一个2x2图像网格，每列都有一个底部的色条。
    """
    grid = AxesGrid(fig, 121,  # 类似于subplot(121)
                    nrows_ncols=(2, 2),
                    axes_pad=0.10,
                    share_all=True,
                    label_mode="1",
                    cbar_location="bottom",
                    cbar_mode="edge",
                    cbar_pad=0.25,
                    cbar_size="15%",
                    direction="column"
                    )

    Z, extent = load_demo_image()
    cmaps = ["autumn", "summer"]
    for i in range(4):
        img = grid[i].imshow(Z, extent=extent, cmap=cmaps[i // 2])
        if i % 2:
            grid.cbar_axes[i // 2].colorbar(img)

    for cax in grid.cbar_axes:
        cax.axis[cax.orientation].set_label("色条标签")

    grid.axes_llc.set_xticks([-2, 0, 2])
    grid.axes_llc.set_yticks([-2, 0, 2])

def create_right_colorbar(fig):
    """
    创建一个2x2图像网格，每行都有一个右侧的色条。
    """
    grid = AxesGrid(fig, 122,  # 类似于subplot(122)
                    nrows_ncols=(2, 2),
                    axes_pad=0.10,
                    label_mode="1",
                    share_all=True,
                    cbar_location="right",
                    cbar_mode="edge",
                    cbar_size="7%",
                    cbar_pad="2%",
                    )
    Z, extent = load_demo_image()
    cmaps = ["spring", "winter"]
    for i in range(4):
        img = grid[i].imshow(Z, extent=extent, cmap=cmaps[i // 2])
        if i % 2:
            grid.cbar_axes[i // 2].colorbar(img)

    for cax in grid.cbar_axes:
        cax.axis[cax.orientation].set_label('色条')

    grid.axes_llc.set_xticks([-2, 0, 2])
    grid.axes_llc.set_yticks([-2, 0, 2])

# 创建图形并绘制两个子图
fig = plt.figure()
create_bottom_colorbar(fig)
create_right_colorbar(fig)

# 显示图形
plt.show()
```
