# 三维散点图

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
n = 100  # 数据点数量

# 三组数据
x1 = np.random.normal(0, 10, n)  # 第一组 x 轴数据
y1 = np.random.normal(0, 10, n)  # 第一组 y 轴数据
z1 = np.random.normal(0, 10, n)  # 第一组 z 轴数据

x2 = np.random.normal(20, 10, n)  # 第二组 x 轴数据
y2 = np.random.normal(20, 10, n)  # 第二组 y 轴数据
z2 = np.random.normal(20, 10, n)  # 第二组 z 轴数据

x3 = np.random.normal(-20, 10, n)  # 第三组 x 轴数据
y3 = np.random.normal(-20, 10, n)  # 第三组 y 轴数据
z3 = np.random.normal(-20, 10, n)  # 第三组 z 轴数据

# 创建画布和 3D 子图
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# 绘制三组三维散点图
ax.scatter(x1, y1, z1, color='#1f77b4', alpha=0.8, edgecolor='black', s=50, label='Group A')
ax.scatter(x2, y2, z2, color='#ff7f0e', alpha=0.8, edgecolor='black', s=50, label='Group B')
ax.scatter(x3, y3, z3, color='#2ca02c', alpha=0.8, edgecolor='black', s=50, label='Group C')

# 设置标题和轴标签
ax.set_xlabel('X 轴', fontsize=12)
ax.set_ylabel('Y 轴', fontsize=12)
ax.set_zlabel('Z 轴', fontsize=12)
ax.set_title('三维散点图', fontsize=16, pad=20)

# 添加图例
ax.legend(loc='upper right', fontsize=12)
ax.grid(False)
# 显示图形
plt.show()
```
