# 绘制分割线分割散点

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
x1 = np.random.normal(0, 10, 100)  # 第一组 x 轴数据
y1 = np.random.normal(50, 15, 100)  # 第一组 y 轴数据

x2 = np.random.normal(0, 10, 100)  # 第二组 x 轴数据
y2 = np.random.normal(50, 15, 100)  # 第二组 y 轴数据

x3 = np.random.normal(0, 10, 100)  # 第三组 x 轴数据
y3 = np.random.normal(50, 15, 100)  # 第三组 y 轴数据

# 创建画布
fig, ax = plt.subplots(figsize=(12, 8))

# 调整 x 轴位置，将三组数据分开
x1_shifted = x1 - 30  # 第一组数据向左移动
x2_shifted = x2       # 第二组数据保持不变
x3_shifted = x3 + 30  # 第三组数据向右移动

# 绘制三组散点图
ax.scatter(x1_shifted, y1, color='#1f77b4', alpha=0.8, edgecolor='black', s=80, label='Group A')
ax.scatter(x2_shifted, y2, color='#ff7f0e', alpha=0.8, edgecolor='black', s=80, label='Group B')
ax.scatter(x3_shifted, y3, color='#2ca02c', alpha=0.8, edgecolor='black', s=80, label='Group C')

# 添加分割线
ax.axvline(x=-15, color='gray', linestyle='--', linewidth=1, alpha=0.7)  # 第一组和第二组之间的分割线
ax.axvline(x=15, color='gray', linestyle='--', linewidth=1, alpha=0.7)   # 第二组和第三组之间的分割线

# 设置标题和轴标签
ax.set_xlabel('X 轴', fontsize=12)
ax.set_ylabel('Y 轴', fontsize=12)
ax.set_title('三组数据散点图', fontsize=16, pad=20)

# 隐藏 x 轴刻度
ax.set_xticks([])

# 添加图例
ax.legend(loc='upper right', fontsize=12)

# 显示图形
plt.savefig('scatter_fenge_1.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
x1 = np.random.normal(0, 10, 100)  # 第一组 x 轴数据
y1 = np.random.normal(50, 15, 100)  # 第一组 y 轴数据

x2 = np.random.normal(0, 10, 100)  # 第二组 x 轴数据
y2 = np.random.normal(50, 15, 100)  # 第二组 y 轴数据

x3 = np.random.normal(0, 10, 100)  # 第三组 x 轴数据
y3 = np.random.normal(50, 15, 100)  # 第三组 y 轴数据

# 创建画布
fig, ax = plt.subplots(figsize=(12, 8))

# 调整 x 轴位置，将三组数据分开
x1_shifted = x1 - 30  # 第一组数据向左移动
x2_shifted = x2       # 第二组数据保持不变
x3_shifted = x3 + 30  # 第三组数据向右移动

# 绘制三组散点图
ax.scatter(x1_shifted, y1, color='#1f77b4', alpha=0.8, edgecolor='black', s=80, label='Group A')
ax.scatter(x2_shifted, y2, color='#ff7f0e', alpha=0.8, edgecolor='black', s=80, label='Group B')
ax.scatter(x3_shifted, y3, color='#2ca02c', alpha=0.8, edgecolor='black', s=80, label='Group C')

# 添加分割线
# 分割线的位置需要根据调整后的 x 轴位置设置
ax.axvline(x=-15, color='gray', linestyle='--', linewidth=1, alpha=0.7)  # 第一组和第二组之间的分割线
ax.axvline(x=15, color='gray', linestyle='--', linewidth=1, alpha=0.7)   # 第二组和第三组之间的分割线

# 设置标题和轴标签
ax.set_xlabel('X 轴', fontsize=12)
ax.set_ylabel('Y 轴', fontsize=12)
ax.set_title('三组数据散点图', fontsize=16, pad=20)

# 隐藏 x 轴刻度
ax.set_xticks([])

# 添加图例
ax.legend(loc='upper right', fontsize=12)

# 显示图形
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
X, y = make_blobs(n_samples=300, centers=3, cluster_std=5.0, random_state=42)

# 将数据集分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 创建 SVM 分类器
clf = svm.SVC(kernel='linear')  # 使用线性核函数
clf.fit(X_train, y_train)  # 训练模型

# 在测试集上进行预测
y_pred = clf.predict(X_test)

# 输出分类报告和混淆矩阵
print("分类报告：")
print(classification_report(y_test, y_pred))

print("混淆矩阵：")
print(confusion_matrix(y_test, y_pred))

# 自定义颜色映射
from matplotlib.colors import ListedColormap
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # 三个类别的颜色
background_color = '#F4E3D7'  # 背景颜色
cmap = ListedColormap(colors)  # 创建自定义颜色映射

# 可视化决策边界
def plot_decision_boundary(X, y, model):
    # 创建网格点
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                         np.arange(y_min, y_max, 0.01))

    # 预测网格点的类别
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # 绘制决策边界
    plt.figure(figsize=(10, 8))
    plt.contourf(xx, yy, Z, alpha=0.8, cmap=cmap)  # 使用自定义颜色映射
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', marker='o', cmap=cmap)
    plt.xlabel('特征 1', fontsize=12)
    plt.ylabel('特征 2', fontsize=12)
    plt.title('SVM 三分类决策边界', fontsize=16, pad=20)
    plt.gca().set_facecolor(background_color)  # 设置背景颜色
    plt.savefig('scatter_svm_three_class.png', dpi=400, bbox_inches='tight')
    plt.show()

# 绘制决策边界
plot_decision_boundary(X_train, y_train, clf)
```
