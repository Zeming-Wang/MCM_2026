# 多组数据+分布直方图

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 设置中文字体和负号显示
plt.rcParams['font.sans-serif'] = 'SimHei'
plt.rcParams['axes.unicode_minus'] = False

# 设置随机种子保证可重复性
np.random.seed(42)

# 生成三组示例数据（二维正态分布）
mean1 = [0, 0]  # 第一组数据的均值
mean2 = [2, 2]  # 第二组数据的均值
mean3 = [-2, -2]  # 第三组数据的均值
cov = [[1, 0.5], [0.5, 1]]  # 协方差矩阵

# 生成三组数据
x1, y1 = np.random.multivariate_normal(mean1, cov, 1000).T
x2, y2 = np.random.multivariate_normal(mean2, cov, 1000).T
x3, y3 = np.random.multivariate_normal(mean3, cov, 1000).T

# 创建画布和子图布局
plt.figure(figsize=(10, 8), facecolor='white')

# 使用Seaborn的JointGrid绘制组合图
g = sns.JointGrid(marginal_ticks=False)  # 隐藏边际图的刻度

# 绘制第一组数据的散点图（主图）
sns.scatterplot(
    x=x1, y=y1,
    ax=g.ax_joint,
    color='#1f77b4',  # 第一组数据颜色
    alpha=0.6,  # 点透明度
    edgecolor='#1f77b4',  # 点边缘颜色
    linewidth=0.3,  # 边缘线宽
    label='Group 1'
)

# 绘制第二组数据的散点图（主图）
sns.scatterplot(
    x=x2, y=y2,
    ax=g.ax_joint,
    color='#ff7f0e',  # 第二组数据颜色
    alpha=0.6,
    edgecolor='#ff7f0e',
    linewidth=0.3,
    label='Group 2'
)

# 绘制第三组数据的散点图（主图）
sns.scatterplot(
    x=x3, y=y3,
    ax=g.ax_joint,
    color='#2ca02c',  # 第三组数据颜色
    alpha=0.6,
    edgecolor='#2ca02c',
    linewidth=0.3,
    label='Group 3'
)

# 绘制第一组数据的边际直方图
sns.histplot(
    x=x1,
    ax=g.ax_marg_x,
    color='#1f77b4',  # 第一组数据颜色
    bins=20,  # 直方图柱数
    kde=True,  # 显示核密度估计曲线
    element='step',  # 直方图样式（阶梯状）
    fill=False  # 不填充颜色
)

# 绘制第二组数据的边际直方图
sns.histplot(
    x=x2,
    ax=g.ax_marg_x,
    color='#ff7f0e',  # 第二组数据颜色
    bins=20,
    kde=True,
    element='step',
    fill=False
)

# 绘制第三组数据的边际直方图
sns.histplot(
    x=x3,
    ax=g.ax_marg_x,
    color='#2ca02c',  # 第三组数据颜色
    bins=20,
    kde=True,
    element='step',
    fill=False
)

# 绘制第一组数据的边际直方图（y轴）
sns.histplot(
    y=y1,
    ax=g.ax_marg_y,
    color='#1f77b4',  # 第一组数据颜色
    bins=20,
    kde=True,
    element='step',
    fill=False
)

# 绘制第二组数据的边际直方图（y轴）
sns.histplot(
    y=y2,
    ax=g.ax_marg_y,
    color='#ff7f0e',  # 第二组数据颜色
    bins=20,
    kde=True,
    element='step',
    fill=False
)

# 绘制第三组数据的边际直方图（y轴）
sns.histplot(
    y=y3,
    ax=g.ax_marg_y,
    color='#2ca02c',  # 第三组数据颜色
    bins=20,
    kde=True,
    element='step',
    fill=False
)

# 添加图形装饰
plt.suptitle('三组二维数据分布及边际分布', y=1.02, fontsize=16)
g.ax_joint.set(
    xlabel='X轴变量',
    ylabel='Y轴变量',
    xlim=(-6, 6),
    ylim=(-6, 6)
)

# 显示图例
g.ax_joint.legend(loc='upper right')

# 调整布局
plt.tight_layout()
plt.savefig('scatter_3_group_bianji.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib.font_manager as fm

# 设置字体路径
font_path = r'../simkai.ttf'  # Windows示例

# 加载字体并设置
font_prop = fm.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False

# 设置随机种子以确保结果可重复
np.random.seed(42)

# 定义颜色
colors = ['#ffaba5', '#1f77b4', '#2ca02c']  # 三组数据的颜色

# 生成三组示例数据
def generate_data():
    # 第一组数据
    x1 = np.random.normal(loc=0, scale=1, size=100)
    y1 = 0.5 * x1 + np.random.normal(loc=0, scale=0.5, size=100)
    
    # 第二组数据
    x2 = np.random.normal(loc=2, scale=1, size=100)
    y2 = 0.5 * x2 + np.random.normal(loc=2, scale=0.5, size=100)
    
    # 第三组数据
    x3 = np.random.normal(loc=-2, scale=1, size=100)
    y3 = 0.5 * x3 + np.random.normal(loc=-2, scale=0.5, size=100)
    
    return x1, y1, x2, y2, x3, y3

# 生成数据
x1, y1, x2, y2, x3, y3 = generate_data()

# 设置图形参数
plt.figure(figsize=(10, 8), constrained_layout=True)  # 设置画布大小
sns.set(style='ticks')  # 使用 Seaborn 的 ticks 风格

# 定义网格布局
grid = plt.GridSpec(4, 4, hspace=0.5, wspace=0.5)  # 4x4网格，设置间距

# 绘制散点图（主图区域）
ax_main = plt.subplot(grid[1:4, 0:3])  # 主图区域

# 绘制第一组数据
sns.scatterplot(
    x=x1, y=y1,
    color=colors[0],  # 第一组数据颜色
    s=50,  # 散点大小
    alpha=0.8,  # 透明度
    label='Group 1'
)

# 绘制第二组数据
sns.scatterplot(
    x=x2, y=y2,
    color=colors[1],  # 第二组数据颜色
    s=50,
    alpha=0.8,
    label='Group 2'
)

# 绘制第三组数据
sns.scatterplot(
    x=x3, y=y3,
    color=colors[2],  # 第三组数据颜色
    s=50,
    alpha=0.8,
    label='Group 3'
)

# 设置主图标题和标签
ax_main.set_title('三组散点图', fontsize=14, fontweight='bold', pad=10, fontproperties=font_prop)
ax_main.set_xlabel('X轴', fontsize=12, labelpad=10, fontproperties=font_prop)
ax_main.set_ylabel('Y轴', fontsize=12, labelpad=10, fontproperties=font_prop)

# 设置坐标轴的颜色和线型
for spine in ax_main.spines.values():
    spine.set_color('#ffaba5')  # 设置坐标轴线颜色为黑色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线
ax_main.tick_params(colors='#ffaba5') # 设置坐标轴颜色

# 绘制x轴数据分布直方图
ax_x_dist = plt.subplot(grid[0, 0:3])  # x轴分布区域

# 绘制第一组数据的x轴分布
sns.histplot(
    x=x1,
    color=colors[0],  # 第一组数据颜色
    kde=True,  # 显示核密度估计曲线
    bins=15,  # 直方图分箱数
    alpha=0.5,  # 透明度
    label='Group 1'
)

# 绘制第二组数据的x轴分布
sns.histplot(
    x=x2,
    color=colors[1],  # 第二组数据颜色
    kde=True,
    bins=15,
    alpha=0.5,
    label='Group 2'
)

# 绘制第三组数据的x轴分布
sns.histplot(
    x=x3,
    color=colors[2],  # 第三组数据颜色
    kde=True,
    bins=15,
    alpha=0.5,
    label='Group 3'
)

# 设置x轴分布图标题和标签
ax_x_dist.set_title('X轴数据分布', fontsize=12, pad=10, fontproperties=font_prop)
ax_x_dist.set_xlabel('', fontproperties=font_prop)  # 隐藏x轴标签
ax_x_dist.set_ylabel('频数', fontsize=10, fontproperties=font_prop)
ax_x_dist.legend(loc='upper right') # 设置图例
ax_x_dist.axis('off') # 关闭坐标轴

# 绘制y轴数据分布直方图
ax_y_dist = plt.subplot(grid[1:4, 3])  # y轴分布区域

# 绘制第一组数据的y轴分布
sns.histplot(
    y=y1,
    color=colors[0],  # 第一组数据颜色
    kde=True,  # 显示核密度估计曲线
    bins=15,  # 直方图分箱数
    alpha=0.5,  # 透明度
    label='Group 1'
)

# 绘制第二组数据的y轴分布
sns.histplot(
    y=y2,
    color=colors[1],  # 第二组数据颜色
    kde=True,
    bins=15,
    alpha=0.5,
    label='Group 2'
)

# 绘制第三组数据的y轴分布
sns.histplot(
    y=y3,
    color=colors[2],  # 第三组数据颜色
    kde=True,
    bins=15,
    alpha=0.5,
    label='Group 3'
)

# 设置y轴分布图标题和标签
ax_y_dist.set_title('Y轴数据分布', fontsize=12, pad=10, fontproperties=font_prop)
ax_y_dist.set_ylabel('', fontproperties=font_prop)  # 隐藏y轴标签
ax_y_dist.set_xlabel('频数', fontsize=10, fontproperties=font_prop)
ax_y_dist.legend(loc='upper right')
ax_y_dist.axis('off') # 关闭坐标轴

# 显示图形
plt.tight_layout()  # 自动调整布局
plt.savefig('scatter_3_groups.png', dpi=400, bbox_inches='tight')  # 保存图形
plt.show()
```
