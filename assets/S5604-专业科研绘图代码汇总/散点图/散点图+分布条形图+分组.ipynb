# 分组散点图

# 分布条形图分开显示

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C']  # 分组名称
colors = ['#ffaba5', '#b6deff', '#ffe587']  # 每组颜色
markers = ['o', 's', 'D']  # 每组标记

# 生成每组的数据
x = []
y = []
for i, group in enumerate(groups):
    x.append(np.random.normal(50 + i * 10, 15, 100))  # x 轴数据
    y.append(np.random.normal(50 + i * 10, 15, 100))  # y 轴数据

# 创建画布
fig = plt.figure(figsize=(14, 10))

# 主散点图
ax_main = fig.add_axes([0.1, 0.3, 0.6, 0.6])  # 手动指定位置和大小

# 绘制分组散点图
for i, group in enumerate(groups):
    ax_main.scatter(x[i], y[i], color=colors[i], marker=markers[i], alpha=0.8, edgecolor=colors[i], s=80, label=group)

# 设置坐标轴的颜色和线型
for spine in ax_main.spines.values():
    spine.set_color('#ffaba5')  # 设置坐标轴线颜色为红色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 2
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 设置刻度线的颜色和线型
ax_main.tick_params(axis='both', which='both', colors='#ffaba5', width=2, length=6, direction='inout')

# 添加图例
ax_main.legend(loc='upper right', fontsize=12)

# 设置标题和轴标签
ax_main.set_xlabel('X 轴', fontsize=12, color='#ffaba5')
ax_main.set_ylabel('Y 轴', fontsize=12, color='#ffaba5')
ax_main.set_title('分组散点图', fontsize=16, pad=20, color='#ffaba5')

# 只显示 x 轴和 y 轴的线，隐藏其他边框线
ax_main.spines['top'].set_visible(False)    # 隐藏上边框
ax_main.spines['right'].set_visible(False)  # 隐藏右边框
ax_main.spines['bottom'].set_visible(True)  # 显示下边框（x 轴）
ax_main.spines['left'].set_visible(True)    # 显示左边框（y 轴）

# 设置坐标轴线的样式
ax_main.spines['bottom'].set_color('#ffaba5')  # 设置 x 轴线颜色
ax_main.spines['left'].set_color('#ffaba5')    # 设置 y 轴线颜色
ax_main.spines['bottom'].set_linewidth(1)    # 设置 x 轴线宽度
ax_main.spines['left'].set_linewidth(1)      # 设置 y 轴线宽度

# X 轴分布条形码（调整到下边）
ax_x_dist = fig.add_axes([0.1, 0.18, 0.6, 0.08])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_x_dist.eventplot(x[i], orientation='horizontal', colors=colors[i], lineoffsets=0.5 + i * 0.2, linelengths=0.2)
ax_x_dist.set_xlim(ax_main.get_xlim())  # 与主图 x 轴范围一致
ax_x_dist.axis('off')  # 隐藏坐标轴

# Y 轴分布条形码（调整到左边，自定义位置）
ax_y_dist = fig.add_axes([0.01, 0.3, 0.05, 0.6])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_y_dist.eventplot(y[i], orientation='vertical', colors=colors[i], lineoffsets=0.5 + i * 0.2, linelengths=0.2)
ax_y_dist.set_ylim(ax_main.get_ylim())  # 与主图 y 轴范围一致
ax_y_dist.axis('off')  # 隐藏坐标轴

# 显示图形
plt.savefig('scatter_group_1.png', dpi=400, bbox_inches='tight')
plt.show()
```

# 条形分布图合并显示

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C']  # 分组名称
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # 每组颜色
markers = ['o', 's', 'D']  # 每组标记

# 生成每组的数据
x = []
y = []
for i, group in enumerate(groups):
    x.append(np.random.normal(50 + i * 10, 15, 100))  # x 轴数据
    y.append(np.random.normal(50 + i * 10, 15, 100))  # y 轴数据

# 创建画布
fig = plt.figure(figsize=(14, 10))

# 主散点图
ax_main = fig.add_axes([0.1, 0.3, 0.6, 0.6])  # 手动指定位置和大小

# 绘制分组散点图
for i, group in enumerate(groups):
    ax_main.scatter(x[i], y[i], color=colors[i], marker=markers[i], alpha=0.8, edgecolor=colors[i], s=80, label=group)

# 设置坐标轴的颜色和线型
for spine in ax_main.spines.values():
    spine.set_color('#1f77b4')  # 设置坐标轴线颜色为红色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 设置刻度线的颜色和线型
ax_main.tick_params(axis='both', which='both', color='#1f77b4', width=2, length=6, direction='inout')

# 添加图例
ax_main.legend(loc='upper right', fontsize=12)

# 设置标题和轴标签
ax_main.set_xlabel('X 轴', fontsize=12, color='#1f77b4')
ax_main.set_ylabel('Y 轴', fontsize=12, color='#1f77b4')
ax_main.set_title('分组散点图', fontsize=16, pad=20, color='#1f77b4')

# X 轴分布条形码（调整到下边，合并绘制）
ax_x_dist = fig.add_axes([0.1, 0.18, 0.6, 0.08])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_x_dist.eventplot(x[i], orientation='horizontal', colors=colors[i], lineoffsets=0.5, linelengths=0.2)
ax_x_dist.set_xlim(ax_main.get_xlim())  # 与主图 x 轴范围一致
ax_x_dist.axis('off')  # 隐藏坐标轴

# Y 轴分布条形码（调整到左边，合并绘制）
ax_y_dist = fig.add_axes([0.01, 0.3, 0.05, 0.6])  # 手动指定位置和大小
for i, group in enumerate(groups):
    ax_y_dist.eventplot(y[i], orientation='vertical', colors=colors[i], lineoffsets=0.5, linelengths=0.2)
ax_y_dist.set_ylim(ax_main.get_ylim())  # 与主图 y 轴范围一致
ax_y_dist.axis('off')  # 隐藏坐标轴

# 显示图形
plt.savefig('scatter_group_2.png', dpi=400, bbox_inches='tight')
plt.show()
```
