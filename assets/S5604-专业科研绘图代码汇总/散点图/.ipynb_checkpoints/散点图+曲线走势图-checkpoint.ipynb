# 散点+曲线走势图

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
x = np.random.normal(50, 15, 100)  # x 轴数据
y = 2 * x + 10 + np.random.normal(0, 10, 100)  # y 轴数据，基于线性关系生成

# 创建画布
fig, ax = plt.subplots(figsize=(10, 6))

# 绘制散点图
ax.scatter(x, y, color='#1f77b4', alpha=0.8, edgecolor='black', s=80, label='数据点')

# 拟合趋势线（线性拟合）
coefficients = np.polyfit(x, y, deg=1)  # 使用一次多项式拟合（线性拟合）
trend_line = np.poly1d(coefficients)  # 创建多项式函数
x_trend = np.linspace(min(x), max(x), 100)  # 生成趋势线的 x 值
y_trend = trend_line(x_trend)  # 计算趋势线的 y 值

# 绘制趋势线
ax.plot(x_trend, y_trend, color='red', linewidth=2, label='趋势线')

# 设置标题和轴标签
ax.set_xlabel('X 轴', fontsize=12)
ax.set_ylabel('Y 轴', fontsize=12)
ax.set_title('散点图与趋势线', fontsize=16, pad=20)

# 添加图例
ax.legend(loc='upper left', fontsize=12)

# 显示网格
# ax.grid(True, linestyle='--', alpha=0.6)
# 隐藏网格
ax.grid(False)

# 显示图形
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
x = np.random.normal(50, 15, 100)  # x 轴数据
y = 2 * x + 10 + np.random.normal(0, 20, 100)  # y 轴数据，基于线性关系生成

# 创建画布
fig, ax = plt.subplots(figsize=(10, 6))

# 绘制散点图
ax.scatter(x, y, color='#1f77b4', alpha=0.8, edgecolor='black', s=80, label='数据点')

# 拟合趋势线（线性）
coefficients_linear = np.polyfit(x, y, deg=1)  # 线性拟合
trend_line_linear = np.poly1d(coefficients_linear)  # 创建线性多项式函数
x_trend = np.linspace(min(x), max(x), 100)  # 生成趋势线的 x 值
y_trend_linear = trend_line_linear(x_trend)  # 计算线性趋势线的 y 值

# 拟合趋势线（二次）
coefficients_quadratic = np.polyfit(x, y, deg=2)  # 二次拟合
trend_line_quadratic = np.poly1d(coefficients_quadratic)  # 创建二次多项式函数
y_trend_quadratic = trend_line_quadratic(x_trend)  # 计算二次趋势线的 y 值

# 拟合趋势线（三次）
coefficients_cubic = np.polyfit(x, y, deg=3)  # 三次拟合
trend_line_cubic = np.poly1d(coefficients_cubic)  # 创建三次多项式函数
y_trend_cubic = trend_line_cubic(x_trend)  # 计算三次趋势线的 y 值

# 绘制趋势线
ax.plot(x_trend, y_trend_linear, color='red', linewidth=2, label='线性趋势线')
ax.plot(x_trend, y_trend_quadratic, color='green', linewidth=2, label='二次趋势线')
ax.plot(x_trend, y_trend_cubic, color='purple', linewidth=2, label='三次趋势线')

# 设置标题和轴标签
ax.set_xlabel('X 轴', fontsize=12)
ax.set_ylabel('Y 轴', fontsize=12)
ax.set_title('散点图与趋势线', fontsize=16, pad=20)

# 添加图例
ax.legend(loc='upper left', fontsize=12)

# 隐藏网格
ax.grid(False)

# 显示图形
plt.savefig('scatter_qu_merge.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'
plt.rcParams['axes.unicode_minus'] = False
# 生成示例数据
np.random.seed(42)
x1 = np.random.normal(50, 15, 100)  # 第一组 x 轴数据
y1 = 2 * x1 + 10 + np.random.normal(0, 20, 100)  # 第一组 y 轴数据

x2 = np.random.normal(70, 15, 100)  # 第二组 x 轴数据
y2 = -0.5 * x2 + 50 + np.random.normal(0, 20, 100)  # 第二组 y 轴数据

x3 = np.random.normal(30, 15, 100)  # 第三组 x 轴数据
y3 = 0.1 * x3**2 + np.random.normal(0, 20, 100)  # 第三组 y 轴数据

# 创建画布和子图
fig, axs = plt.subplots(3, 1, figsize=(10, 15))

# 第一组数据：线性趋势
axs[0].scatter(x1, y1, color='#1f77b4', alpha=0.8, edgecolor='black', s=80, label='数据点')
coefficients_linear = np.polyfit(x1, y1, deg=1)  # 线性拟合
trend_line_linear = np.poly1d(coefficients_linear)  # 创建线性多项式函数
x_trend = np.linspace(min(x1), max(x1), 100)  # 生成趋势线的 x 值
y_trend_linear = trend_line_linear(x_trend)  # 计算线性趋势线的 y 值
axs[0].plot(x_trend, y_trend_linear, color='red', linewidth=2, label='线性趋势线')
axs[0].set_title('第一组数据：线性趋势', fontsize=14)
axs[0].legend(loc='upper left', fontsize=12)

# 第二组数据：线性趋势（负斜率）
axs[1].scatter(x2, y2, color='#ff7f0e', alpha=0.8, edgecolor='black', s=80, label='数据点')
coefficients_linear2 = np.polyfit(x2, y2, deg=1)  # 线性拟合
trend_line_linear2 = np.poly1d(coefficients_linear2)  # 创建线性多项式函数
x_trend2 = np.linspace(min(x2), max(x2), 100)  # 生成趋势线的 x 值
y_trend_linear2 = trend_line_linear2(x_trend2)  # 计算线性趋势线的 y 值
axs[1].plot(x_trend2, y_trend_linear2, color='green', linewidth=2, label='线性趋势线')
axs[1].set_title('第二组数据：线性趋势（负斜率）', fontsize=14)
axs[1].legend(loc='upper left', fontsize=12)

# 第三组数据：二次趋势
axs[2].scatter(x3, y3, color='#2ca02c', alpha=0.8, edgecolor='black', s=80, label='数据点')
coefficients_quadratic = np.polyfit(x3, y3, deg=2)  # 二次拟合
trend_line_quadratic = np.poly1d(coefficients_quadratic)  # 创建二次多项式函数
x_trend3 = np.linspace(min(x3), max(x3), 100)  # 生成趋势线的 x 值
y_trend_quadratic = trend_line_quadratic(x_trend3)  # 计算二次趋势线的 y 值
axs[2].plot(x_trend3, y_trend_quadratic, color='purple', linewidth=2, label='二次趋势线')
axs[2].set_title('第三组数据：二次趋势', fontsize=14)
axs[2].legend(loc='upper left', fontsize=12)

# 设置全局标题
fig.suptitle('三组数据散点图与趋势线', fontsize=16, y=1.02)

# 调整子图间距
plt.tight_layout()

# 显示图形
plt.savefig('scatter_qu.png', dpi=400, bbox_inches='tight')
plt.show()
```
