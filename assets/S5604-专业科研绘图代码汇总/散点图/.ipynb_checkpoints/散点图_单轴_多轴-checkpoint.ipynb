# 单轴散点图

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
data = np.random.normal(50, 15, 100)  # 生成 100 个正态分布的数据点

# 创建画布
fig, ax = plt.subplots(figsize=(10, 6))

# 绘制单轴散点图（沿 x 轴分布）
ax.scatter(
    data,  # x 轴数据
    np.zeros_like(data),  # y 轴数据（全部为 0）
    color='#1f77b4',  # 数据点颜色
    alpha=0.8,  # 透明度
    edgecolor='black',  # 边缘颜色
    s=100  # 数据点大小
)

# 隐藏 y 轴
ax.get_yaxis().set_visible(False)

# 设置标题和轴标签
ax.set_xlabel('数据值', fontsize=12, color='#1f77b4')
ax.set_title('单轴散点图（沿 X 轴分布）', fontsize=16, pad=20, color='#1f77b4')

# 设置坐标轴的颜色和线型
for spine in ax.spines.values():
    spine.set_color('#1f77b4')  # 设置坐标轴线颜色为蓝色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 显示图形
plt.savefig('single_axis_scatter.png', dpi=400, bbox_inches='tight')
plt.show()
```

# 多轴散点图

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C', 'Group D', 'Group E', 'Group F']  # 六组名称
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']  # 每组颜色

# 生成每组的数据
data = []
for i in range(len(groups)):
    data.append(np.random.normal(50 + 5 * np.random.randint(1, 5), 10, 20))  # 每组生成 20 个正态分布的数据点

# 创建画布
fig, ax = plt.subplots(figsize=(10, 8))

# 绘制六组单轴散点图
for i, group in enumerate(groups):
    ax.scatter(
        data[i],  # x 轴数据
        np.full_like(data[i], i),  # y 轴数据（每组数据沿不同的 y 轴位置分布）
        color=colors[i],  # 数据点颜色
        alpha=0.8,  # 透明度
        edgecolor=colors[i],  # 边缘颜色
        s=100,  # 数据点大小
        label=group  # 分组标签
    )

# 设置 y 轴刻度和标签
ax.set_yticks(range(len(groups)))  # 设置 y 轴刻度
ax.set_yticklabels(groups, fontsize=12)  # 设置 y 轴标签

# 设置标题和轴标签
ax.set_xlabel('数据值', fontsize=12, color='#1f77b4')
ax.set_ylabel('分组', fontsize=12, color='#1f77b4')
ax.set_title('六组单轴散点图', fontsize=16, pad=20, color='#1f77b4')

# 设置坐标轴的颜色和线型
for spine in ax.spines.values():
    spine.set_color('#1f77b4')  # 设置坐标轴线颜色为蓝色
    spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
    spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 添加图例
ax.legend(loc='upper right', fontsize=12)

# 显示图形
plt.savefig('six_group_single_axis_scatter.png', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体
plt.rcParams['font.family'] = 'SimHei'

# 生成示例数据
np.random.seed(42)
groups = ['Group A', 'Group B', 'Group C', 'Group D', 'Group E', 'Group F']  # 六组名称
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']  # 每组颜色

# 生成每组的数据
data = []
for i in range(len(groups)):
    data.append(np.random.normal(50 + 5 * np.random.randint(1, 5), 10, 20))  # 每组生成 20 个正态分布的数据点

# 创建画布
fig, ax = plt.subplots(figsize=(10, 8))

# 绘制六组单轴散点图
for i, group in enumerate(groups):
    ax.scatter(
        data[i],  # x 轴数据
        np.full_like(data[i], i),  # y 轴数据（每组数据沿不同的 y 轴位置分布）
        color=colors[i],  # 数据点颜色
        alpha=0.8,  # 透明度
        edgecolor=colors[i],  # 边缘颜色
        s=100,  # 数据点大小
        label=group  # 分组标签
    )

# 在每个轴位置绘制一条虚线
for i in range(len(groups)):
    ax.axhline(
        y=i,  # y 轴位置
        color='gray',  # 虚线颜色
        linestyle='--',  # 虚线样式
        linewidth=1,  # 线宽
        alpha=0.5  # 透明度
    )

# 设置 y 轴刻度和标签
ax.set_yticks(range(len(groups)))  # 设置 y 轴刻度
ax.set_yticklabels(groups, fontsize=12)  # 设置 y 轴标签

# 设置标题和轴标签
ax.set_xlabel('数据值', fontsize=12, color='#1f77b4')
ax.set_ylabel('分组', fontsize=12, color='#1f77b4')
ax.set_title('六组单轴散点图（带虚线）', fontsize=16, pad=20, color='#1f77b4')

# 设置坐标轴的颜色和线型
# for spine in ax.spines.values():
#     spine.set_color('#1f77b4')  # 设置坐标轴线颜色为蓝色
#     spine.set_linewidth(1)  # 设置坐标轴线宽度为 1
#     spine.set_linestyle('--')  # 设置坐标轴线型为虚线

# 只显示 x 轴和 y 轴的线，隐藏其他边框线
ax.spines['top'].set_visible(False)    # 隐藏上边框
ax.spines['right'].set_visible(False)  # 隐藏右边框
ax.spines['bottom'].set_visible(True)  # 显示下边框（x 轴）
ax.spines['bottom'].set_color('#ffaba5')  # 显示下边框（x 轴）
ax.spines['left'].set_visible(False)    # 隐藏左边框（y 轴）
ax.tick_params(colors='#ffaba5')
# 添加图例
ax.legend(loc='upper right', fontsize=12)

# 显示图形
plt.savefig('six_group_single_axis_scatter_with_lines.png', dpi=400, bbox_inches='tight')
plt.show()
```
